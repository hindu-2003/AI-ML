### Decision Tree Parameter


criterion: Tells the tree how to measure the “quality” of a split.

        gini → measures impurity (default)
        
        entropy → uses information gain
        
        log_loss → probability-based impurity


max_depth : Maximum levels the tree can grow.

        If None → tree grows until all leaves are pure (can overfit).
        
        Small value (like 3–10) → reduces overfitting.

Example: max_depth=3 means tree will have 3 levels only.



min_samples_split: Minimum number of samples needed to split a node.

        Default: 2
        
        If you set min_samples_split=10, the node must have at least 10 samples to split.

When to increase?

        When your model is overfitting.



min_samples_leaf: Minimum number of samples required in a leaf node.

    Default: 1
    
    If you set min_samples_leaf=5, every final leaf must have at least 5 samples.

Why important?

    Prevents leaves with very few samples → more generalization.



max_features: How many features to consider when looking for the best split.

Options:

    "auto" / "sqrt" → good for classification
    
    "log2"
    
    None → use all features



splitter: How the tree chooses splits.
    
    "best" → finds the best split (default)
    
    "random" → chooses a random split (useful for ensemble diversity)

Only matters when building a single tree or a randomized tree.



max_leaf_nodes : Maximum number of leaf nodes allowed.
    
    If None → unlimited
    
    If set → tree will grow only until it reaches that many leaves

Good for restricting complexity.
