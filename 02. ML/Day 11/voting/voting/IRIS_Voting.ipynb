### Voting Classifier Iris Data

```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split, cross_val_score

from sklearn.preprocessing import StandardScaler

from sklearn.linear_model import LogisticRegression

from sklearn.tree import DecisionTreeClassifier

from sklearn.svm import SVC

from sklearn.neighbors import KNeighborsClassifier

from sklearn.ensemble import VotingClassifier

from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
```

```python
data1=pd.read_csv('Iris_Data.csv')
data1
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
data1['class'] = le.fit_transform(data1['species'])
data1
```

```python
x=data1.iloc[:,0:4]
y=data1.iloc[:,5]
```

```python
x
```

```python
y
```

```python
minmax=MinMaxScaler(feature_range=(0,1))

x = minmax.fit_transform(x)
```

```python
x
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.3)
```

```python
clf1 = LogisticRegression(random_state=42, max_iter=1000)
clf2 = DecisionTreeClassifier(random_state=42, max_depth=4)
clf3 = KNeighborsClassifier(n_neighbors=5)
clf4 = SVC(probability=True, random_state=42, kernel='rbf')

classifiers = [
    ('Logistic Regression', clf1),
    ('Decision Tree', clf2),
    ('K-Nearest Neighbors', clf3),
    ('Support Vector Machine', clf4)
]
```

```python
classifiers
```

```python
individual_results = {}

for name, clf in classifiers:
    # Train
    clf.fit(x_train, y_train)
    
    # Predict
    y_pred = clf.predict(x_test)
    
    # Evaluate
    acc = accuracy_score(y_test, y_pred)
    cv_scores = cross_val_score(clf, x_train, y_train, cv=5)
    
    individual_results[name] = {
        'accuracy': acc,
        'cv_mean': cv_scores.mean(),
        'cv_std': cv_scores.std()
    }
    
    print(f"\n{name}:")
    print(f"  Test Accuracy: {acc:.4f} ({acc*100:.2f}%)")
    print(f"  CV Accuracy: {cv_scores.mean():.4f} (±{cv_scores.std():.4f})")
```

```python

voting_hard = VotingClassifier(
    estimators=[
        ('lr', clf1),
        ('dt', clf2),
        ('knn', clf3),
        ('svm', clf4)
    ],
    voting='hard'
)

voting_hard.fit(x_train, y_train)
y_pred_hard = voting_hard.predict(x_test)
acc_hard = accuracy_score(y_test, y_pred_hard)
cv_scores_hard = cross_val_score(voting_hard, x_train, y_train, cv=5)

print(f"\nHard Voting Performance:")
print(f"  Test Accuracy: {acc_hard:.4f} ({acc_hard*100:.2f}%)")
print(f"  CV Accuracy: {cv_scores_hard.mean():.4f} (±{cv_scores_hard.std():.4f})")
```

```python

voting_soft = VotingClassifier(
    estimators=[
        ('lr', clf1),
        ('dt', clf2),
        ('knn', clf3),
        ('svm', clf4)
    ],
    voting='soft'
)

voting_soft.fit(x_train, y_train)
y_pred_soft = voting_soft.predict(x_test)
acc_soft = accuracy_score(y_test, y_pred_soft)
cv_scores_soft = cross_val_score(voting_soft, x_train, y_train, cv=5)

print(f"\nSoft Voting Performance:")
print(f"  Test Accuracy: {acc_soft:.4f} ({acc_soft*100:.2f}%)")
print(f"  CV Accuracy: {cv_scores_soft.mean():.4f} (±{cv_scores_soft.std():.4f})")
```
