```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.metrics import accuracy_score
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import VotingClassifier
```

```python
# Load Data

df = pd.read_csv("heart.csv")
df
```

```python
# Detect numeric & categorical columns

numeric_cols = df.select_dtypes(include=['int64','float64']).columns.tolist()

categorical_cols = df.select_dtypes(include=['object']).columns.tolist()
```

```python
numeric_cols 
```

```python
categorical_cols
```

```python
# Remove target from lists

numeric_cols.remove("HeartDisease")
```

```python

# Boxplot BEFORE capping

plt.figure(figsize=(15,5))
sns.boxplot(data=df[numeric_cols])
plt.title("Boxplot of Numeric Features BEFORE Capping")
plt.show()
```

```python

# Outlier Capping Function (IQR/Winsorization)

df_capped = df.copy()
df_capped[numeric_cols] = df_capped[numeric_cols].astype(float)

def cap_outliers_iqr(series):
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    series = series.copy()
    series[series < lower] = float(lower)
    series[series > upper] = float(upper)
    return series

for col in numeric_cols:
    df_capped[col] = cap_outliers_iqr(df_capped[col])

# Boxplot AFTER capping

plt.figure(figsize=(15,5))
sns.boxplot(data=df_capped[numeric_cols])
plt.title("Boxplot of Numeric Features AFTER Capping")
plt.show()
```

```python

# Handle Missing Values

df_capped[numeric_cols] = df_capped[numeric_cols].fillna(df_capped[numeric_cols].median())
df_capped[categorical_cols] = df_capped[categorical_cols].fillna(df_capped[categorical_cols].mode().iloc[0])
```

```python
# Encode Categorical

le = LabelEncoder()
for col in categorical_cols:
    df_capped[col] = le.fit_transform(df_capped[col])
```

```python
# Split X and y

X = df_capped.drop("HeartDisease", axis=1)
y = df_capped["HeartDisease"]
```

```python
X
```

```python

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python

# Scale numeric features

scaler = StandardScaler()
X_train[numeric_cols] = scaler.fit_transform(X_train[numeric_cols])
X_test[numeric_cols] = scaler.transform(X_test[numeric_cols])
```

```python

# Voting Classifier (SVM + Logistic Regression)

model_lr = LogisticRegression(max_iter=300)
model_svm = SVC(kernel='rbf', probability=True)

voting = VotingClassifier(
    estimators=[('lr', model_lr), ('svm', model_svm)],
    voting='soft'
)
```

```python

voting.fit(X_train, y_train)
pred = voting.predict(X_test)

print("Accuracy:", accuracy_score(y_test, pred))
```

```python
voting = VotingClassifier(
    estimators=[('lr', model_lr), ('svm', model_svm)],
    voting='hard'
)

voting.fit(X_train, y_train)
pred = voting.predict(X_test)

print("Accuracy (Hard Voting):", accuracy_score(y_test, pred))
```
