```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import DBSCAN

# Step 1: Load Iris dataset
iris = load_iris()
X = iris.data  # Using only features (no target labels)

# Step 2: Standardize the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Step 3: Apply DBSCAN
db = DBSCAN(eps=0.6, min_samples=5)
labels = db.fit_predict(X_scaled)

# Step 4: Add results to a DataFrame for inspection
df = pd.DataFrame(X, columns=iris.feature_names)
df['Cluster'] = labels

print(df.head())

# Step 5: Visualize the clusters using two principal features
plt.figure(figsize=(8, 6))
plt.scatter(X_scaled[:, 0], X_scaled[:, 1], c=labels, cmap='rainbow')
plt.title("DBSCAN Clustering on Iris Dataset")
plt.xlabel("Standardized Sepal Length")
plt.ylabel("Standardized Sepal Width")
plt.grid(True)
plt.show()
```
