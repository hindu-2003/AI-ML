
Hierarchical clustering used to group, similar data points into clusters based on their distance (or similarity).

Unlike K-Means, you don’t need to predefine the number of clusters — 
      
      can visualize them in a dendrogram and cut it at a desired level.
      

Two Types of Hierarchical Clustering

    Agglomerative (Bottom-Up) → Start with each data point as its own cluster and merge step-by-step.
     Most common approach.
    
    Divisive (Top-Down) → Start with all data points in one cluster and split recursively.

<b>Steps in Agglomerative Hierarchical Clustering</b>

Compute the distance matrix between every object to other object

    Usually Euclidean distance between all pairs of points.

Merge closest clusters

    Based on a linkage criterion .

Update the distance matrix

    Recalculate distances between the new cluster and others.

Repeat

    Until all points are merged into a single cluster.

Visualize with a Dendrogram

    A tree-like diagram showing how clusters were merged.
    
    Cutting the dendrogram at a chosen level gives the desired number of clusters.
    

### Linkage Criteria

| Linkage Type         | Description                                  | Effect                         |
| -------------------- | -------------------------------------------- | ------------------------------ |
| **Single Linkage**   | Min distance between points in two clusters  | Can create “chained” clusters  |
| **Complete Linkage** | Max distance between points in two clusters  | Produces compact clusters      |
| **Average Linkage**  | Average distance between all pairs of points | Balanced clustering            |
| **Ward’s Method**    | Minimizes variance within clusters           | Most common for numerical data |
