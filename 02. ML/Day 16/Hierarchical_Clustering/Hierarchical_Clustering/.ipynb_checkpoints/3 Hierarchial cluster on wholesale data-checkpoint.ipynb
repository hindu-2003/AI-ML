```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python

data = pd.read_csv('Wholesale customers data.csv')
data
```

```python

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

data_scaled = pd.DataFrame(scaler.fit_transform(data), columns=data.columns)
data_scaled.head()
```

```python

import scipy.cluster.hierarchy as shc

plt.figure(figsize=(10, 7))  
plt.title("Dendrograms")  

dend = shc.dendrogram(shc.linkage(data_scaled, method='ward'))
plt.show()
```

```python

plt.figure(figsize=(10, 7))  
plt.title("Dendrograms")  

dend = shc.dendrogram(shc.linkage(data_scaled, method='ward'))
plt.axhline(y=29, color='r', linestyle='--')
plt.show()
```

```python

from sklearn.cluster import AgglomerativeClustering

cluster = AgglomerativeClustering(n_clusters=2, linkage='ward')  
cluster.fit_predict(data_scaled)
```

```python

plt.figure(figsize=(10, 7))  

plt.scatter(data_scaled['Milk'], data_scaled['Grocery'], c=cluster.labels_) 
plt.show()
```

```python

from sklearn.metrics import silhouette_score

scores = []
for k in range(2, 10):   #  cluster counts from 2 to 9
    model = AgglomerativeClustering(n_clusters=k, linkage='ward')
    labels = model.fit_predict(data_scaled)
    score = silhouette_score(data_scaled, labels)
    scores.append((k, score))

for k, s in scores:
    print(f"Clusters: {k}, Silhouette Score: {s:.3f}")
```

```python

plt.plot([k for k, s in scores], [s for k, s in scores], marker='o')

plt.xlabel("Number of clusters (k)")
plt.ylabel("Silhouette Score")
plt.title("Choosing Optimal k")
plt.show()
```
