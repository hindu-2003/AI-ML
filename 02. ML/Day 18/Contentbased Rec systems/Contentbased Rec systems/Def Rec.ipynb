A content-based recommendation system recommends items based on how similar they are in terms of their features or attributes.

| Type                           | Description                                                                       | Example                                                      |
| ------------------------------ | --------------------------------------------------------------------------------- | ------------------------------------------------------------ |
| **1. Content-Based Filtering** | Recommends items similar to those a user liked before, based on **item features** | If you like **Mango**, suggest fruits with similar nutrients |
| **2. Collaborative Filtering** | Recommends items based on **similar users’ preferences**                          | If users who like Mango also like Banana, recommend Banana   |
| **3. Hybrid Models**           | Combines both content and collaborative filtering                                 | Netflix uses this                                            |
