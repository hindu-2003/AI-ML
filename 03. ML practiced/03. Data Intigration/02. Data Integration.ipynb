# Data Integration in Machine Learning 🔗

## What is Data Integration?

**Data Integration** is the process of combining data from multiple sources into a single, unified dataset that can be used for analysis or machine learning.

### Why Do We Need Data Integration?
- Data comes from different sources (databases, CSV files, APIs, etc.)
- Different formats and structures
- Need clean, consistent data for ML models
- Avoid data redundancy and inconsistencies

## Types of Data Integration

1. **Horizontal Integration** (Combining columns): Merge datasets side-by-side
2. **Vertical Integration** (Combining rows): Stack datasets on top of each other
3. **Transformation Integration**: Convert data to common format
4. **Aggregation Integration**: Combine data at different levels

````python
import pandas as pd
import numpy as np

# ============================================================================
# EXAMPLE 1: HORIZONTAL INTEGRATION (Side-by-side merge)
# ============================================================================
print("=" * 70)
print("EXAMPLE 1: HORIZONTAL INTEGRATION (Adding Columns)")
print("=" * 70)

# Imagine you have customer data from two different sources
# Source 1: Customer names and ages
customer_data = {
    'Customer_ID': [1, 2, 3, 4],
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 28]
}
df1 = pd.DataFrame(customer_data)

print("\n📊 Source 1 - Customer Demographics:")
print(df1)

# Source 2: Customer purchase history
purchase_data = {
    'Customer_ID': [1, 2, 3, 4],
    'Purchase_Amount': [500, 1200, 800, 450],
    'Items_Bought': [3, 5, 2, 1]
}
df2 = pd.DataFrame(purchase_data)

print("\n🛒 Source 2 - Purchase History:")
print(df2)

# Merge them horizontally on Customer_ID
merged_horizontal = pd.merge(df1, df2, on='Customer_ID')

print("\n✅ After Horizontal Integration (MERGED):")
print(merged_horizontal)
print("\nNow we have complete customer profile: Demographics + Purchase History!")
````

````python
# ============================================================================
# EXAMPLE 2: VERTICAL INTEGRATION (Stacking rows)
# ============================================================================
print("\n" + "=" * 70)
print("EXAMPLE 2: VERTICAL INTEGRATION (Combining Data from Same Source)")
print("=" * 70)

# Data from January
jan_sales = {
    'Date': ['Jan 1', 'Jan 2', 'Jan 3'],
    'Product': ['Laptop', 'Mouse', 'Keyboard'],
    'Sales': [1200, 50, 80]
}
df_jan = pd.DataFrame(jan_sales)

print("\n📅 January Sales Data:")
print(df_jan)

# Data from February
feb_sales = {
    'Date': ['Feb 1', 'Feb 2', 'Feb 3'],
    'Product': ['Monitor', 'Headset', 'Laptop'],
    'Sales': [350, 120, 1200]
}
df_feb = pd.DataFrame(feb_sales)

print("\n📅 February Sales Data:")
print(df_feb)

# Stack them vertically
combined_sales = pd.concat([df_jan, df_feb], ignore_index=True)

print("\n✅ After Vertical Integration (COMBINED):")
print(combined_sales)
print("\nNow we have all sales data for analysis across months!")
````

````python
# ============================================================================
# EXAMPLE 3: DATA STANDARDIZATION (Format Conversion)
# ============================================================================
print("\n" + "=" * 70)
print("EXAMPLE 3: DATA STANDARDIZATION (Format Conversion)")
print("=" * 70)

# Database format (SQL)
db_data = {
    'ID': [101, 102, 103],
    'SALARY': [50000, 60000, 55000],
    'DEPT_CODE': ['ENG', 'HR', 'SALES']
}
df_db = pd.DataFrame(db_data)

print("\n🗄️ Data from Database (Original Format):")
print(df_db)
print(f"Data types:\n{df_db.dtypes}")

# Standardize: lowercase column names, consistent data types
df_standardized = df_db.copy()
df_standardized.columns = df_standardized.columns.str.lower()
df_standardized['salary'] = df_standardized['salary'].astype(float)

# Map department codes to full names
dept_mapping = {'ENG': 'Engineering', 'HR': 'Human Resources', 'SALES': 'Sales'}
df_standardized['department'] = df_standardized['dept_code'].map(dept_mapping)
df_standardized = df_standardized.drop('dept_code', axis=1)

print("\n✅ After Standardization (Normalized Format):")
print(df_standardized)
print(f"Data types:\n{df_standardized.dtypes}")
print("\nNow the data is consistent and ready for ML models!")
````

````python
# ============================================================================
# EXAMPLE 4: COMPLETE DATA INTEGRATION PIPELINE
# ============================================================================
print("\n" + "=" * 70)
print("EXAMPLE 4: COMPLETE DATA INTEGRATION PIPELINE")
print("=" * 70)

# Step 1: Load data from multiple sources
print("\n🔍 Step 1: Loading Data from Multiple Sources")
print("-" * 70)

# Source A: Online Store Data
online_data = {
    'Order_ID': [1001, 1002, 1003, 1004],
    'Customer_ID': [101, 102, 101, 103],
    'Order_Amount': [250, 500, 150, 800],
    'Order_Date': ['2024-01-01', '2024-01-02', '2024-01-03', '2024-01-04']
}
df_online = pd.DataFrame(online_data)
print("📱 Online Store Orders:")
print(df_online)

# Source B: Physical Store Data
store_data = {
    'Order_ID': [2001, 2002, 2003],
    'Customer_ID': [101, 104, 102],
    'Order_Amount': [100, 350, 200],
    'Order_Date': ['2024-01-01', '2024-01-02', '2024-01-04']
}
df_store = pd.DataFrame(store_data)
print("\n🏪 Physical Store Orders:")
print(df_store)

# Source C: Customer Profile
customer_data = {
    'Customer_ID': [101, 102, 103, 104],
    'Name': ['John', 'Sarah', 'Mike', 'Lisa'],
    'Email': ['john@email.com', 'sarah@email.com', 'mike@email.com', 'lisa@email.com']
}
df_customer = pd.DataFrame(customer_data)
print("\n👤 Customer Profiles:")
print(df_customer)

# Step 2: Combine vertical data (Online + Store)
print("\n\n📊 Step 2: Combining Orders (Vertical Integration)")
print("-" * 70)
df_all_orders = pd.concat([df_online, df_store], ignore_index=True)
print("✅ All Orders Combined:")
print(df_all_orders)

# Step 3: Merge with customer information (Horizontal Integration)
print("\n\n🔗 Step 3: Adding Customer Information (Horizontal Integration)")
print("-" * 70)
df_integrated = pd.merge(df_all_orders, df_customer, on='Customer_ID')
print("✅ Complete Integrated Dataset:")
print(df_integrated)

# Step 4: Data Cleanup and Enhancement
print("\n\n🧹 Step 4: Data Cleanup and Feature Engineering")
print("-" * 70)
df_final = df_integrated.copy()

# Convert Order_Date to datetime
df_final['Order_Date'] = pd.to_datetime(df_final['Order_Date'])

# Create new features
df_final['Month'] = df_final['Order_Date'].dt.month
df_final['Day_of_Week'] = df_final['Order_Date'].dt.day_name()

# Categorize order amount
def categorize_amount(amount):
    if amount < 200:
        return 'Small'
    elif amount < 500:
        return 'Medium'
    else:
        return 'Large'

df_final['Order_Category'] = df_final['Order_Amount'].apply(categorize_amount)

print("✅ Final Integrated & Enriched Dataset:")
print(df_final)

print("\n✨ Summary:")
print(f"   • Total Orders: {len(df_final)}")
print(f"   • Total Customers: {df_final['Customer_ID'].nunique()}")
print(f"   • Total Revenue: ${df_final['Order_Amount'].sum()}")
print(f"   • Average Order Value: ${df_final['Order_Amount'].mean():.2f}")
````

## Key Functions for Data Integration

| Function | Purpose | Example |
|----------|---------|---------|
| `pd.merge()` | Join dataframes on common column | `pd.merge(df1, df2, on='ID')` |
| `pd.concat()` | Stack dataframes (vertical/horizontal) | `pd.concat([df1, df2])` |
| `pd.join()` | Join on index | `df1.join(df2)` |
| `pd.melt()` | Unpivot data | `df.melt(id_vars=['ID'])` |
| `pd.pivot()` | Pivot data | `df.pivot(index='ID', columns='Month')` |

````python
# ============================================================================
# EXAMPLE 5: DIFFERENT TYPES OF MERGES
# ============================================================================
print("\n" + "=" * 70)
print("EXAMPLE 5: DIFFERENT TYPES OF JOINS/MERGES")
print("=" * 70)

# Create sample dataframes
left_df = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Salary': [50000, 60000, 75000, 55000]
})

right_df = pd.DataFrame({
    'ID': [1, 2, 4, 5],
    'Department': ['IT', 'HR', 'Sales', 'Marketing'],
    'Manager': ['Tom', 'Jane', 'Bob', 'Alice']
})

print("\n👤 Left DataFrame (Employees):")
print(left_df)
print("\n🏢 Right DataFrame (Departments):")
print(right_df)

# 1. INNER JOIN (Only matching records)
print("\n📌 1. INNER JOIN - Keep only matching records:")
inner = pd.merge(left_df, right_df, on='ID', how='inner')
print(inner)
print("→ Only IDs 1, 2, 4 appear (all are in both datasets)")

# 2. LEFT JOIN (All from left, matching from right)
print("\n📌 2. LEFT JOIN - Keep all left records:")
left_join = pd.merge(left_df, right_df, on='ID', how='left')
print(left_join)
print("→ All 4 employees shown; ID 3 has NaN for Department")

# 3. RIGHT JOIN (All from right, matching from left)
print("\n📌 3. RIGHT JOIN - Keep all right records:")
right_join = pd.merge(left_df, right_df, on='ID', how='right')
print(right_join)
print("→ All departments shown; ID 5 has NaN for Name/Salary")

# 4. OUTER JOIN (All records from both)
print("\n📌 4. OUTER JOIN - Keep all records from both:")
outer = pd.merge(left_df, right_df, on='ID', how='outer')
print(outer)
print("→ All 5 IDs appear; NaN where data doesn't match")
````

````python
# ============================================================================
# EXAMPLE 6: REAL-WORLD SCENARIO - E-COMMERCE DATA INTEGRATION
# ============================================================================
print("\n" + "=" * 70)
print("EXAMPLE 6: REAL-WORLD - E-COMMERCE DATA INTEGRATION")
print("=" * 70)

print("\n🎯 Scenario: We need to analyze which products are most profitable")
print("   We have data from 3 sources that need to be combined\n")

# Source 1: Product Information
products = {
    'Product_ID': [1, 2, 3, 4],
    'Product_Name': ['Laptop', 'Mouse', 'Keyboard', 'Monitor'],
    'Cost_Price': [500, 10, 30, 200]
}
df_products = pd.DataFrame(products)
print("📦 Source 1 - Products:")
print(df_products)

# Source 2: Sales Data
sales = {
    'Sale_ID': [101, 102, 103, 104, 105],
    'Product_ID': [1, 1, 2, 3, 1],
    'Selling_Price': [800, 800, 20, 50, 850],
    'Quantity': [2, 1, 5, 3, 1]
}
df_sales = pd.DataFrame(sales)
print("\n💰 Source 2 - Sales:")
print(df_sales)

# Source 3: Customer Reviews
reviews = {
    'Sale_ID': [101, 102, 103, 104, 105],
    'Rating': [5, 4, 3, 5, 4],
    'Review': ['Excellent', 'Good', 'Average', 'Great', 'Good']
}
df_reviews = pd.DataFrame(reviews)
print("\n⭐ Source 3 - Customer Reviews:")
print(df_reviews)

# Step 1: Merge Sales with Products
print("\n\n🔗 Step 1: Merge Sales with Products")
print("-" * 70)
df_merged1 = pd.merge(df_sales, df_products, on='Product_ID')
print(df_merged1)

# Step 2: Add Reviews
print("\n\n🔗 Step 2: Add Customer Reviews")
print("-" * 70)
df_final_data = pd.merge(df_merged1, df_reviews, on='Sale_ID')
print(df_final_data)

# Step 3: Calculate Profitability
print("\n\n📊 Step 3: Calculate Profitability Metrics")
print("-" * 70)
df_final_data['Total_Revenue'] = df_final_data['Selling_Price'] * df_final_data['Quantity']
df_final_data['Total_Cost'] = df_final_data['Cost_Price'] * df_final_data['Quantity']
df_final_data['Profit'] = df_final_data['Total_Revenue'] - df_final_data['Total_Cost']
df_final_data['Profit_Margin_%'] = (df_final_data['Profit'] / df_final_data['Total_Revenue'] * 100).round(2)

print(df_final_data[['Product_Name', 'Quantity', 'Total_Revenue', 'Profit', 'Profit_Margin_%', 'Rating']])

# Step 4: Summary Analysis
print("\n\n📈 Step 4: Business Insights")
print("-" * 70)
profit_by_product = df_final_data.groupby('Product_Name').agg({
    'Profit': 'sum',
    'Profit_Margin_%': 'mean',
    'Rating': 'mean'
}).round(2)

print(profit_by_product)
print("\n🏆 Best performing product (by profit):", profit_by_product['Profit'].idxmax())
print("⭐ Best rated product:", profit_by_product['Rating'].idxmax())
````

## Common Data Integration Challenges & Solutions

### Challenge 1: Duplicate Records
**Problem:** Same data appears multiple times
```
ID | Name   | Email
1  | John   | john@email.com
1  | John   | john@email.com  ← Duplicate!
```
**Solution:** Use `drop_duplicates()`
```python
df_clean = df.drop_duplicates(subset=['ID'])
```

### Challenge 2: Missing Values (NaN)
**Problem:** Some fields are empty after merge
**Solution:** Use `fillna()` or drop rows with NaN
```python
df.fillna(value=0)  # Replace with 0
df.dropna()         # Remove rows with NaN
```

### Challenge 3: Data Type Mismatches
**Problem:** Same column has different types (string vs number)
**Solution:** Convert data types
```python
df['Age'] = df['Age'].astype(int)
```

### Challenge 4: Inconsistent Values
**Problem:** Department names are different ('HR' vs 'hr' vs 'Human Resources')
**Solution:** Standardize values
```python
df['Department'] = df['Department'].str.upper()
df['Department'] = df['Department'].map({'HR': 'Human Resources'})
```

````python
# ============================================================================
# EXAMPLE 7: HANDLING INTEGRATION CHALLENGES
# ============================================================================
print("\n" + "=" * 70)
print("EXAMPLE 7: HANDLING DATA INTEGRATION CHALLENGES")
print("=" * 70)

# Challenge 1: Duplicates
print("\n❌ Challenge 1: Duplicate Records")
print("-" * 70)
messy_df = pd.DataFrame({
    'ID': [1, 1, 2, 3, 3],
    'Name': ['John', 'John', 'Sarah', 'Mike', 'Mike'],
    'Score': [85, 85, 90, 75, 75]
})
print("Before removing duplicates:")
print(messy_df)

clean_df = messy_df.drop_duplicates()
print("\n✅ After removing duplicates:")
print(clean_df)

# Challenge 2: Missing Values
print("\n\n❌ Challenge 2: Missing Values (NaN)")
print("-" * 70)
df_missing = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Name': ['John', 'Sarah', None, 'Mike'],
    'Age': [25, None, 30, 28],
    'Salary': [50000, 60000, None, 55000]
})
print("Before handling missing values:")
print(df_missing)
print(f"\nMissing values count:\n{df_missing.isnull().sum()}")

# Solution 1: Drop rows with any NaN
df_dropped = df_missing.dropna()
print("\n✅ Solution 1 - Drop rows with missing values:")
print(df_dropped)

# Solution 2: Fill missing values
df_filled = df_missing.copy()
df_filled['Name'] = df_filled['Name'].fillna('Unknown')
df_filled['Age'] = df_filled['Age'].fillna(df_filled['Age'].mean())
df_filled['Salary'] = df_filled['Salary'].fillna(df_filled['Salary'].median())
print("\n✅ Solution 2 - Fill missing values:")
print(df_filled)

# Challenge 3: Data Type Issues
print("\n\n❌ Challenge 3: Wrong Data Types")
print("-" * 70)
df_wrong_types = pd.DataFrame({
    'ID': ['1', '2', '3'],  # String, should be int
    'Age': ['25', '30', '28'],  # String, should be int
    'Salary': [50000, 60000, 55000]  # Number, correct
})
print("Before type conversion:")
print(df_wrong_types)
print(f"\nData types:\n{df_wrong_types.dtypes}")

df_correct_types = df_wrong_types.copy()
df_correct_types['ID'] = df_correct_types['ID'].astype(int)
df_correct_types['Age'] = df_correct_types['Age'].astype(int)

print("\n✅ After type conversion:")
print(df_correct_types)
print(f"\nData types:\n{df_correct_types.dtypes}")

# Challenge 4: Inconsistent Values
print("\n\n❌ Challenge 4: Inconsistent Categorical Values")
print("-" * 70)
df_inconsistent = pd.DataFrame({
    'ID': [1, 2, 3, 4, 5],
    'Department': ['it', 'IT', 'Engineering', 'eng', 'It'],
    'Status': ['active', 'ACTIVE', 'inactive', 'Active', 'inactive']
})
print("Before standardization:")
print(df_inconsistent)

df_standardized = df_inconsistent.copy()

# Standardize Department
df_standardized['Department'] = df_standardized['Department'].str.lower()
dept_mapping = {'it': 'IT', 'engineering': 'Engineering', 'eng': 'Engineering', 'hr': 'HR'}
df_standardized['Department'] = df_standardized['Department'].map(dept_mapping)

# Standardize Status
df_standardized['Status'] = df_standardized['Status'].str.lower()

print("\n✅ After standardization:")
print(df_standardized)
````

## Quick Reference: Data Integration Checklist

Before using integrated data for ML:

- ✅ **Load** - Import data from all sources
- ✅ **Explore** - Check shape, columns, data types
- ✅ **Combine** - Merge/concat data appropriately
- ✅ **Clean** - Remove duplicates and handle missing values
- ✅ **Standardize** - Convert formats and values to be consistent
- ✅ **Validate** - Check for errors and inconsistencies
- ✅ **Feature Engineer** - Create new useful columns
- ✅ **Ready** - Data is now ready for ML models!

## Summary

- **Data Integration** combines data from multiple sources
- **Horizontal Integration**: Merge side-by-side (columns)
- **Vertical Integration**: Stack top-to-bottom (rows)
- **Standardization**: Convert to common format
- **Clean & Validate**: Remove issues before modeling
- **Real-world use**: E-commerce, finance, healthcare, etc.
