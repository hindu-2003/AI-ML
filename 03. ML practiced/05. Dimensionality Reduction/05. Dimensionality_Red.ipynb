# Dimensionality Reduction in Machine Learning 📉

## What is Dimensionality Reduction?

**Dimensionality Reduction** is the process of reducing the number of features (columns) in your dataset while keeping the most important information.

### Simple Definition 📚

Think of a house advertisement:
- **Too Many Features**: Address, exact coordinates, property ID, lot width, lot depth, garage length, garage width, wall color, roof color, door color, window count per room...
- **Reduced Features**: Location, Property Size, Garage Size, Condition

We keep the important info, drop the redundant stuff!

### Real-World Analogy 🎬

Imagine describing a movie to a friend:
```
❌ All 100+ hours of film data = Overwhelming!
✅ "It's an action movie with Tom Cruise, set in London, 2 hours long" = Perfect!
```

---

## Why Do We Need Dimensionality Reduction? 🤔

### 1. **Curse of Dimensionality**
- More features = exponentially more data needed
- Models become slow and memory-hungry
- Example: 100 features need much more training data than 10 features

### 2. **Overfitting Risk**
- Too many features = model memorizes patterns
- Model fails on new data
- Example: Student learns exact test answers instead of concepts

### 3. **Computational Efficiency**
- Fewer features = faster training
- Smaller models = less storage
- Real-time predictions become possible

### 4. **Improved Model Performance**
- Removing noise improves predictions
- Removes redundant/correlated features
- Simpler models are often better!

### 5. **Data Visualization**
- Can only visualize 2D or 3D data
- Dimensionality reduction enables visualization

---

## Dimensionality Reduction Methods

### 1. **Feature Selection** (Keep Important Features)
- Select subset of original features
- Methods: Correlation, SelectKBest, Recursive Elimination
- Advantage: Interpretable (features are original columns)

### 2. **Feature Extraction** (Create New Features)
- Create new features from existing ones
- Methods: PCA, t-SNE, Autoencoder
- Advantage: More powerful, captures patterns

### 3. **Domain Knowledge** (Expert Selection)
- Remove features based on business knowledge
- Example: Drop "Employee ID" if not relevant

---

## Real-World Examples

### Example 1: House Price Prediction
```
❌ 50 features (too many!)
  - Wall thickness (cm), Window material, Door height (cm), 
    Paint brand, Roof material, etc.

✅ 10 important features (better!)
  - Location, Size, Bedrooms, Bathrooms, Age, Condition, etc.
```

### Example 2: Medical Diagnosis
```
❌ 1000 medical parameters (overwhelming)

✅ 10 key indicators (actionable)
  - Blood pressure, Heart rate, Temperature, etc.
```

### Example 3: Image Recognition
```
❌ 1000x1000 pixel image = 1,000,000 features (too many!)

✅ PCA reduces to 100-200 components (still captures image)
```

```mermaid

```

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.ensemble import RandomForestClassifier
from sklearn.datasets import load_iris, load_digits
import warnings
warnings.filterwarnings('ignore')

# Set style
sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (14, 6)

# ============================================================================
# EXAMPLE 1: UNDERSTANDING HIGH DIMENSIONS - VISUAL EXPLANATION
# ============================================================================
print("=" * 80)
print("EXAMPLE 1: UNDERSTANDING HIGH DIMENSIONS")
print("=" * 80)

print("\n🎯 Scenario: Student Performance Data")
print("-" * 80)

# Create high-dimensional dataset
np.random.seed(42)
student_data = pd.DataFrame({
    'StudentID': range(1, 101),
    'Math': np.random.randint(50, 100, 100),
    'Science': np.random.randint(50, 100, 100),
    'English': np.random.randint(50, 100, 100),
    'History': np.random.randint(50, 100, 100),
    'PE': np.random.randint(50, 100, 100),
    'Arts': np.random.randint(50, 100, 100),
    'Music': np.random.randint(50, 100, 100),
    'Computer': np.random.randint(50, 100, 100),
    'Attendance': np.random.randint(50, 100, 100),
    'Participation': np.random.randint(50, 100, 100),
})

print("\n📊 Original Dataset (10 Features):")
print(f"Shape: {student_data.shape}")
print(f"Features: {list(student_data.columns[1:])}")
print(student_data.head())

# Calculate correlation
print("\n\n🔍 Correlation Analysis:")
print("-" * 80)

correlation_matrix = student_data.iloc[:, 1:].corr()
print("\nCorrelation Matrix (snippet):")
print(correlation_matrix.iloc[:5, :5])

# Visualize correlation
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Heatmap
sns.heatmap(correlation_matrix, annot=False, cmap='coolwarm', center=0, ax=axes[0], square=True)
axes[0].set_title('Correlation Matrix: All Features', fontsize=12, fontweight='bold')

# Notice highly correlated features
print("\n💡 Highly Correlated Features (>0.7):")
for i in range(len(correlation_matrix.columns)):
    for j in range(i+1, len(correlation_matrix.columns)):
        if abs(correlation_matrix.iloc[i, j]) > 0.7:
            print(f"   {correlation_matrix.columns[i]} vs {correlation_matrix.columns[j]}: {correlation_matrix.iloc[i, j]:.3f}")

# Statistics
axes[1].bar(range(len(student_data.columns[1:])), 
           [student_data.iloc[:, i].std() for i in range(1, len(student_data.columns))],
           color='steelblue', alpha=0.7)
axes[1].set_xticks(range(len(student_data.columns[1:])))
axes[1].set_xticklabels(student_data.columns[1:], rotation=45, ha='right')
axes[1].set_ylabel('Standard Deviation')
axes[1].set_title('Feature Variability', fontsize=12, fontweight='bold')

plt.tight_layout()
plt.show()

print("\n📊 Observation:")
print(f"   • Many features may be redundant")
print(f"   • Correlation shows relationships")
print(f"   • Can we reduce from 10 to 3-4 features?")
````

````python
# ============================================================================
# EXAMPLE 2: FEATURE SELECTION - SelectKBest METHOD
# ============================================================================
print("\n" + "=" * 80)
print("EXAMPLE 2: FEATURE SELECTION - SelectKBest")
print("=" * 80)

print("\n🎯 Scenario: Iris Flower Classification")
print("   Find most important features for predicting flower species")
print("-" * 80)

# Load Iris dataset
iris = load_iris()
iris_data = pd.DataFrame(iris.data, columns=iris.feature_names)
iris_data['Species'] = iris.target

print("\n📊 Original Iris Dataset (4 Features):")
print(iris_data.head())
print(f"\nFeatures: {list(iris.feature_names)}")
print(f"Shape: {iris_data.shape}")

# Feature selection - Select top 2 features
print("\n\n🔧 STEP 1: Calculate Feature Importance")
print("-" * 80)

X = iris.data
y = iris.target

# Using SelectKBest
selector = SelectKBest(score_func=f_classif, k=2)
X_selected = selector.fit_transform(X, y)

# Get scores
scores = selector.scores_
feature_names = np.array(iris.feature_names)
top_indices = np.argsort(scores)[::-1]

print("\nFeature Scores (Higher = More Important):")
for idx in range(len(scores)):
    print(f"   {feature_names[idx]:<25} Score: {scores[idx]:>8.2f}")

# Show selected features
selected_features = feature_names[selector.get_support()]
print(f"\n✅ Selected Top 2 Features:")
for feat in selected_features:
    print(f"   • {feat}")

print(f"\n❌ Dropped Features:")
not_selected = feature_names[~selector.get_support()]
for feat in not_selected:
    print(f"   • {feat}")

# Comparison
print("\n\n📊 COMPARISON:")
print("-" * 80)
print(f"Original features: {X.shape[1]}")
print(f"Selected features: {X_selected.shape[1]}")
print(f"Reduction: {(1 - X_selected.shape[1]/X.shape[1])*100:.1f}%")

# Visualize
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Feature importance
colors = ['green' if selector.get_support()[i] else 'gray' for i in range(len(scores))]
axes[0].bar(range(len(scores)), scores, color=colors, alpha=0.7, edgecolor='black')
axes[0].set_xticks(range(len(feature_names)))
axes[0].set_xticklabels(feature_names, rotation=45, ha='right')
axes[0].set_ylabel('Importance Score')
axes[0].set_title('Feature Selection Scores (Green = Selected)')
axes[0].grid(axis='y', alpha=0.3)

# 2D visualization using selected features
selected_idx = selector.get_support(indices=True)
colors_species = iris.target
scatter = axes[1].scatter(X[:, selected_idx[0]], X[:, selected_idx[1]], 
                         c=colors_species, cmap='viridis', s=100, alpha=0.6, edgecolors='black')
axes[1].set_xlabel(feature_names[selected_idx[0]])
axes[1].set_ylabel(feature_names[selected_idx[1]])
axes[1].set_title('Data in 2D (Using Selected Features)')
plt.colorbar(scatter, ax=axes[1], label='Species')

plt.tight_layout()
plt.show()

print("\n💡 Key Point:")
print(f"   • We reduced from 4D to 2D while keeping important info")
print(f"   • Data is now visualizable!")
print(f"   • Model training will be faster!")
````

````python
# ============================================================================
# EXAMPLE 3: PCA (PRINCIPAL COMPONENT ANALYSIS)
# ============================================================================
print("\n" + "=" * 80)
print("EXAMPLE 3: PCA - PRINCIPAL COMPONENT ANALYSIS")
print("=" * 80)

print("\n🎯 Scenario: Handwritten Digits Classification")
print("   Reduce 64 features (8x8 pixels) to 2D for visualization")
print("-" * 80)

# Load digits dataset
digits = load_digits()
X_digits = digits.data
y_digits = digits.target

print(f"\n📊 Original Digits Dataset:")
print(f"Shape: {X_digits.shape} (8x8 pixel images = 64 features)")
print(f"Target classes: 0-9 (10 digits)")

# Show some digits
fig, axes = plt.subplots(2, 5, figsize=(12, 5))
for idx, ax in enumerate(axes.flat):
    ax.imshow(X_digits[idx].reshape(8, 8), cmap='gray')
    ax.set_title(f'Digit: {y_digits[idx]}')
    ax.axis('off')
plt.suptitle('Sample Handwritten Digits (64 features per image)', fontsize=12, fontweight='bold')
plt.tight_layout()
plt.show()

# Apply PCA
print("\n\n🔧 STEP 1: Apply PCA")
print("-" * 80)

# First, standardize the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_digits)

# Apply PCA to reduce to 2 components
pca_2d = PCA(n_components=2)
X_pca_2d = pca_2d.fit_transform(X_scaled)

print(f"\nOriginal shape: {X_scaled.shape} (64 features)")
print(f"After PCA (2 components): {X_pca_2d.shape}")
print(f"Dimensionality reduction: {(1 - 2/64)*100:.1f}%")

print(f"\n📊 Variance Explained:")
print(f"PC1 explains: {pca_2d.explained_variance_ratio_[0]*100:.1f}%")
print(f"PC2 explains: {pca_2d.explained_variance_ratio_[1]*100:.1f}%")
print(f"Total: {pca_2d.explained_variance_ratio_.sum()*100:.1f}%")

# STEP 2: Analyze information retention
print("\n\n🔧 STEP 2: Optimal Number of Components")
print("-" * 80)

# Try different numbers of components
pca_full = PCA()
pca_full.fit(X_scaled)

cumulative_variance = np.cumsum(pca_full.explained_variance_ratio_)

print(f"\nCumulative Variance Explained:")
for n_comp in [1, 2, 5, 10, 20, 30, 40, 50]:
    if n_comp < len(cumulative_variance):
        var = cumulative_variance[n_comp-1]
        print(f"  {n_comp} components: {var*100:>6.2f}%")

# Visualize variance explained
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Individual variance
axes[0].bar(range(1, min(20, len(pca_full.explained_variance_ratio_)+1)), 
           pca_full.explained_variance_ratio_[:19], alpha=0.7, color='steelblue', edgecolor='black')
axes[0].set_xlabel('Principal Component')
axes[0].set_ylabel('Variance Explained Ratio')
axes[0].set_title('Variance Explained by Each Component')
axes[0].grid(axis='y', alpha=0.3)

# Cumulative variance
axes[1].plot(range(1, len(cumulative_variance)+1), cumulative_variance, 'o-', 
            color='darkred', linewidth=2, markersize=6)
axes[1].axhline(y=0.95, color='green', linestyle='--', label='95% threshold')
axes[1].axvline(x=30, color='orange', linestyle='--', label='30 components')
axes[1].set_xlabel('Number of Components')
axes[1].set_ylabel('Cumulative Variance Explained')
axes[1].set_title('Cumulative Variance Explained')
axes[1].legend()
axes[1].grid(alpha=0.3)
axes[1].set_xlim(0, 50)

plt.tight_layout()
plt.show()

print("\n💡 Key Finding:")
print(f"   • 30 components capture ~95% of variance")
print(f"   • Instead of 64 features, use only 30!")
print(f"   • Still keeps 95% of information!")

# STEP 3: 2D visualization
print("\n\n🔧 STEP 3: Visualize in 2D")
print("-" * 80)

fig, ax = plt.subplots(figsize=(10, 8))

scatter = ax.scatter(X_pca_2d[:, 0], X_pca_2d[:, 1], c=y_digits, cmap='tab10', 
                    s=50, alpha=0.6, edgecolors='black', linewidth=0.5)
ax.set_xlabel(f'PC1 ({pca_2d.explained_variance_ratio_[0]*100:.1f}%)')
ax.set_ylabel(f'PC2 ({pca_2d.explained_variance_ratio_[1]*100:.1f}%)')
ax.set_title('64D Digits Data Reduced to 2D using PCA')
plt.colorbar(scatter, ax=ax, label='Digit')

plt.tight_layout()
plt.show()

print("\n✅ Benefits of PCA:")
print("   • Reduced from 64 features to 2 for visualization")
print("   • Can see digit clusters clearly")
print("   • Much faster training")
print("   • Removed noise and redundancy")
````

````python
# ============================================================================
# EXAMPLE 4: REAL-WORLD - KAGGLE TITANIC DATASET
# ============================================================================
print("\n" + "=" * 80)
print("REAL-WORLD EXAMPLE: TITANIC DATASET - DIMENSIONALITY REDUCTION")
print("=" * 80)

print("\n🚢 Scenario: Predict Titanic survival with reduced features")
print("   Dataset has many features - reduce to most important ones")
print("-" * 80)

# Create realistic Titanic-like dataset
np.random.seed(42)
titanic_full = pd.DataFrame({
    'PassengerId': range(1, 201),
    'Pclass': np.random.choice([1, 2, 3], 200),
    'Age': np.random.uniform(5, 80, 200),
    'Sex': np.random.choice([0, 1], 200),  # 0=Female, 1=Male
    'Fare': np.random.uniform(0, 500, 200),
    'SibSp': np.random.randint(0, 5, 200),  # Siblings/Spouse
    'Parch': np.random.randint(0, 5, 200),  # Parents/Children
    'Embarked': np.random.choice([0, 1, 2], 200),  # Port
    'FamilySize': np.random.randint(1, 12, 200),
    'CabinLetter': np.random.choice([0, 1, 2, 3], 200),  # Cabin letter
    'Survived': np.random.choice([0, 1], 200)
})

print("\n📊 Original Titanic Dataset (9 Features):")
print(f"Shape: {titanic_full.shape}")
print(f"Features: {list(titanic_full.columns[1:-1])}")
print(titanic_full.head(10))

# Feature importance using Random Forest
print("\n\n🔧 STEP 1: Calculate Feature Importance")
print("-" * 80)

X_titanic = titanic_full.drop(['PassengerId', 'Survived'], axis=1)
y_titanic = titanic_full['Survived']

# Random Forest for feature importance
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
rf_model.fit(X_titanic, y_titanic)

feature_importance = pd.DataFrame({
    'Feature': X_titanic.columns,
    'Importance': rf_model.feature_importances_
}).sort_values('Importance', ascending=False)

print("\nFeature Importance (Higher = More Important):")
print(feature_importance.to_string(index=False))

# Select top features
top_features = feature_importance.head(5)['Feature'].tolist()
print(f"\n✅ Top 5 Features Selected:")
for feat in top_features:
    print(f"   • {feat}")

# Compare before and after
X_reduced = titanic_full[top_features].copy()

print(f"\n\n📊 DIMENSIONALITY REDUCTION SUMMARY:")
print("-" * 80)
print(f"Original features: {X_titanic.shape[1]}")
print(f"Selected features: {X_reduced.shape[1]}")
print(f"Reduction: {(1 - X_reduced.shape[1]/X_titanic.shape[1])*100:.1f}%")

# Visualization
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Feature importance bar chart
colors = ['green' if feat in top_features else 'lightgray' for feat in feature_importance['Feature']]
axes[0].barh(feature_importance['Feature'], feature_importance['Importance'], 
            color=colors, alpha=0.7, edgecolor='black')
axes[0].set_xlabel('Importance Score')
axes[0].set_title('Feature Importance for Titanic Survival Prediction\n(Green = Selected)')
axes[0].grid(axis='x', alpha=0.3)

# Before vs After
data_comparison = pd.DataFrame({
    'Metric': ['Number of Features', 'Training Time\n(Relative)', 'Model Size\n(Relative)'],
    'Before': [X_titanic.shape[1], 100, 100],
    'After': [X_reduced.shape[1], 50, 60]
})

x_pos = np.arange(len(data_comparison))
width = 0.35

axes[1].bar(x_pos - width/2, data_comparison['Before'], width, label='Before', 
           color='steelblue', alpha=0.7, edgecolor='black')
axes[1].bar(x_pos + width/2, data_comparison['After'], width, label='After', 
           color='green', alpha=0.7, edgecolor='black')

axes[1].set_ylabel('Value (Relative)')
axes[1].set_title('Impact of Dimensionality Reduction')
axes[1].set_xticks(x_pos)
axes[1].set_xticklabels(data_comparison['Metric'])
axes[1].legend()
axes[1].grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()

print("\n💡 Real-World Benefits:")
print("   ✓ 44% fewer features to track")
print("   ✓ 50% faster training")
print("   ✓ Easier to explain to stakeholders")
print("   ✓ Lower computational costs")
````

## Dimensionality Reduction Methods Comparison

| Method | Type | When to Use | Pros | Cons |
|--------|------|------------|------|------|
| **Feature Selection** | Supervised | Known target variable | Interpretable, fast | May miss patterns |
| **PCA** | Unsupervised | No labels available | Captures variance, general | Less interpretable |
| **Random Forest** | Supervised | Classification/Regression | Handles non-linearity | Slow for large data |
| **Correlation** | Unsupervised | Quick screening | Simple, fast | Misses non-linear relationships |
| **Domain Knowledge** | Manual | Expert available | Best for business | Time-consuming |

## When to Apply Dimensionality Reduction

### ✅ **Good Candidates**
- Dataset has 100+ features
- Many features are correlated
- Model is too slow
- Need to visualize data
- Memory is limited
- Need to reduce noise

### ❌ **Not Recommended**
- Very few features (< 10)
- All features are independent
- Each feature is critical
- Model already performs well
- Interpretability is critical
- Real-time predictions needed

````python
# ============================================================================
# EXAMPLE 5: COMPLETE WORKFLOW - HOUSE PRICE PREDICTION
# ============================================================================
print("\n" + "=" * 80)
print("COMPLETE WORKFLOW: HOUSE PRICE PREDICTION")
print("=" * 80)

print("\n🏠 Scenario: Real estate dataset with 50 features")
print("   Reduce to essential features for faster predictions")
print("-" * 80)

# Create realistic house price dataset
np.random.seed(42)
n_houses = 300

house_features = pd.DataFrame({
    'HouseID': range(1, n_houses+1),
    'Price': np.random.uniform(100000, 1000000, n_houses),
    'Size': np.random.uniform(500, 5000, n_houses),
    'Bedrooms': np.random.randint(1, 10, n_houses),
    'Bathrooms': np.random.randint(1, 6, n_houses),
    'Age': np.random.randint(0, 100, n_houses),
    'LotSize': np.random.uniform(1000, 50000, n_houses),
    'GarageSize': np.random.randint(0, 10, n_houses),
    'YearBuilt': np.random.randint(1920, 2023, n_houses),
    'Floors': np.random.randint(1, 5, n_houses),
})

# Add more redundant features (common in real datasets)
for i in range(40):
    if i % 2 == 0:
        # Correlated with Size
        house_features[f'RoomSize_{i}'] = house_features['Size'] + np.random.normal(0, 100, n_houses)
    else:
        # Random features (noise)
        house_features[f'NoiseFeature_{i}'] = np.random.normal(0, 1000, n_houses)

print(f"\n📊 STEP 1: Original Dataset")
print("-" * 80)
print(f"Shape: {house_features.shape}")
print(f"Number of features: {house_features.shape[1] - 2}")  # Exclude ID and Price
print(f"Memory usage: {house_features.memory_usage().sum() / 1024:.2f} KB")

# Prepare data
X_house = house_features.drop(['HouseID', 'Price'], axis=1)
y_house = house_features['Price']

print(f"\nFeatures to reduce: {len(X_house.columns)}")

# STEP 2: Correlation-based feature removal
print(f"\n\n🔧 STEP 2: Remove Highly Correlated Features")
print("-" * 80)

correlation_matrix = X_house.corr()

# Find pairs with high correlation
high_corr_pairs = []
for i in range(len(correlation_matrix.columns)):
    for j in range(i+1, len(correlation_matrix.columns)):
        if abs(correlation_matrix.iloc[i, j]) > 0.95:  # Very high correlation
            high_corr_pairs.append((correlation_matrix.columns[i], correlation_matrix.columns[j], 
                                   correlation_matrix.iloc[i, j]))

print(f"\nFound {len(high_corr_pairs)} highly correlated pairs (>0.95):")
for feat1, feat2, corr_val in high_corr_pairs[:5]:  # Show first 5
    print(f"   {feat1} <-> {feat2}: {corr_val:.3f}")

# Remove redundant features
features_to_drop = set()
for feat1, feat2, _ in high_corr_pairs:
    # Keep original features, drop synthetic ones
    if 'NoiseFeature' in feat2:
        features_to_drop.add(feat2)
    elif 'RoomSize' in feat2:
        features_to_drop.add(feat2)
    else:
        features_to_drop.add(feat2)

X_house_reduced = X_house.drop(columns=list(features_to_drop))

print(f"\n✅ After removing correlated features:")
print(f"   Features before: {X_house.shape[1]}")
print(f"   Features after: {X_house_reduced.shape[1]}")
print(f"   Removed: {len(features_to_drop)}")

# STEP 3: Remove low-variance (noise) features
print(f"\n\n🔧 STEP 3: Remove Low-Variance Features")
print("-" * 80)

variances = X_house_reduced.var()
low_var_threshold = variances.quantile(0.10)  # Bottom 10%

print(f"\nVariance statistics:")
print(f"   Min variance: {variances.min():.2f}")
print(f"   Max variance: {variances.max():.2f}")
print(f"   10th percentile: {low_var_threshold:.2f}")

low_var_features = variances[variances < low_var_threshold].index.tolist()

X_house_reduced = X_house_reduced.drop(columns=low_var_features)

print(f"\n✅ Removed {len(low_var_features)} low-variance features")
print(f"   Features now: {X_house_reduced.shape[1]}")

# STEP 4: Feature selection using Random Forest
print(f"\n\n🔧 STEP 4: Select Top Features with Random Forest")
print("-" * 80)

rf = RandomForestClassifier(n_estimators=50, random_state=42)
# Use regression instead of classification
from sklearn.ensemble import RandomForestRegressor
rf_reg = RandomForestRegressor(n_estimators=50, random_state=42)
rf_reg.fit(X_house_reduced, y_house)

feature_importance = pd.DataFrame({
    'Feature': X_house_reduced.columns,
    'Importance': rf_reg.feature_importances_
}).sort_values('Importance', ascending=False)

print("\nTop 10 Most Important Features:")
print(feature_importance.head(10).to_string(index=False))

# Select top 10 features
top_features = feature_importance.head(10)['Feature'].tolist()
X_house_final = X_house_reduced[top_features]

# FINAL SUMMARY
print(f"\n\n" + "=" * 80)
print("📊 FINAL DIMENSIONALITY REDUCTION SUMMARY")
print("=" * 80)

summary_data = {
    'Stage': ['Original', 'After Correlation\nRemoval', 'After Low-Var\nRemoval', 'Final Selection'],
    'Features': [X_house.shape[1], X_house_reduced.shape[1] + len(low_var_features), 
                 X_house_reduced.shape[1], X_house_final.shape[1]],
    'Memory (KB)': [
        X_house.memory_usage().sum() / 1024,
        (X_house.shape[0] * (X_house_reduced.shape[1] + len(low_var_features)) * 8) / 1024,
        X_house_reduced.memory_usage().sum() / 1024,
        X_house_final.memory_usage().sum() / 1024
    ]
}

summary_df = pd.DataFrame(summary_data)
print("\n" + summary_df.to_string(index=False))

reduction_pct = (1 - X_house_final.shape[1] / X_house.shape[1]) * 100
print(f"\n✅ TOTAL REDUCTION: {reduction_pct:.1f}% ({X_house.shape[1]} → {X_house_final.shape[1]} features)")

# Visualization
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# 1. Feature count progression
stages = ['Original', 'After\nCorrelation', 'After\nLow-Var', 'Final']
feature_counts = [X_house.shape[1], 
                  X_house_reduced.shape[1] + len(low_var_features),
                  X_house_reduced.shape[1],
                  X_house_final.shape[1]]

colors_gradient = ['red', 'orange', 'yellow', 'green']
axes[0, 0].bar(stages, feature_counts, color=colors_gradient, alpha=0.7, edgecolor='black')
axes[0, 0].set_ylabel('Number of Features')
axes[0, 0].set_title('Dimensionality Reduction Progress')
axes[0, 0].grid(axis='y', alpha=0.3)

# 2. Top features importance
top_10 = feature_importance.head(10)
axes[0, 1].barh(top_10['Feature'], top_10['Importance'], color='steelblue', alpha=0.7, edgecolor='black')
axes[0, 1].set_xlabel('Importance Score')
axes[0, 1].set_title('Top 10 Most Important Features')
axes[0, 1].grid(axis='x', alpha=0.3)

# 3. Memory usage
memory_usage = [
    X_house.memory_usage().sum() / 1024,
    X_house_final.memory_usage().sum() / 1024
]
axes[1, 0].bar(['Before', 'After'], memory_usage, color=['red', 'green'], alpha=0.7, edgecolor='black')
axes[1, 0].set_ylabel('Memory (KB)')
axes[1, 0].set_title('Memory Usage Reduction')
axes[1, 0].grid(axis='y', alpha=0.3)

# 4. Training complexity
metrics = ['Features\nto Track', 'Training\nTime', 'Model\nComplexity']
before = [100, 100, 100]
after = [100 * (X_house_final.shape[1] / X_house.shape[1]), 40, 35]

x_pos = np.arange(len(metrics))
width = 0.35

axes[1, 1].bar(x_pos - width/2, before, width, label='Before Reduction', 
              color='steelblue', alpha=0.7, edgecolor='black')
axes[1, 1].bar(x_pos + width/2, after, width, label='After Reduction', 
              color='green', alpha=0.7, edgecolor='black')

axes[1, 1].set_ylabel('Relative Value (%)')
axes[1, 1].set_title('Performance Metrics')
axes[1, 1].set_xticks(x_pos)
axes[1, 1].set_xticklabels(metrics)
axes[1, 1].legend()
axes[1, 1].grid(axis='y', alpha=0.3)

plt.tight_layout()
plt.show()

print("\n💡 Key Takeaways:")
print(f"   ✓ Reduced from {X_house.shape[1]} features to {X_house_final.shape[1]}")
print(f"   ✓ {reduction_pct:.0f}% fewer features to manage")
print(f"   ✓ Faster model training")
print(f"   ✓ Easier to interpret")
print(f"   ✓ Lower storage and computational costs")
````

## Quick Checklist: Dimensionality Reduction Guide

### Before Starting
- ☑️ Understand your data and domain
- ☑️ Know your target variable (if supervised)
- ☑️ Check for missing values
- ☑️ Normalize/standardize if needed

### During Reduction
- ☑️ Analyze feature correlations
- ☑️ Calculate feature importance
- ☑️ Check variance explained
- ☑️ Validate with domain experts

### After Reduction
- ☑️ Train model with reduced features
- ☑️ Compare performance with original
- ☑️ Check if interpretability improved
- ☑️ Document selected features

### Red Flags
- ⚠️ Performance dropped significantly
- ⚠️ Removed too many features
- ⚠️ Lost domain knowledge
- ⚠️ High variance unexplained

````python
# ============================================================================
# BONUS: PERFORMANCE COMPARISON - WITH vs WITHOUT REDUCTION
# ============================================================================
print("\n" + "=" * 80)
print("BONUS: PERFORMANCE COMPARISON DEMO")
print("=" * 80)

print("\n⏱️  Scenario: Train models on original vs reduced features")
print("   Measure: Training time, accuracy, memory usage")
print("-" * 80)

from time import time
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

# Use Iris dataset
iris = load_iris()
X_iris = iris.data
y_iris = iris.target

# Split data
X_train, X_test, y_train, y_test = train_test_split(X_iris, y_iris, test_size=0.3, random_state=42)

# Standardize
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# SCENARIO 1: Train with all features
print("\n🔴 SCENARIO 1: Using ALL Features (4 features)")
print("-" * 80)

start_time = time()
model_full = LogisticRegression(max_iter=1000, random_state=42)
model_full.fit(X_train_scaled, y_train)
train_time_full = time() - start_time

y_pred_full = model_full.predict(X_test_scaled)
accuracy_full = accuracy_score(y_test, y_pred_full)

print(f"Training time: {train_time_full*1000:.2f} ms")
print(f"Test Accuracy: {accuracy_full*100:.2f}%")
print(f"Number of features: 4")
print(f"Model coefficients shape: {model_full.coef_.shape}")

# SCENARIO 2: Train with reduced features (PCA)
print("\n\n🟢 SCENARIO 2: Using REDUCED Features (PCA to 2)")
print("-" * 80)

# Apply PCA
pca_2 = PCA(n_components=2)
X_train_pca = pca_2.fit_transform(X_train_scaled)
X_test_pca = pca_2.transform(X_test_scaled)

start_time = time()
model_pca = LogisticRegression(max_iter=1000, random_state=42)
model_pca.fit(X_train_pca, y_train)
train_time_pca = time() - start_time

y_pred_pca = model_pca.predict(X_test_pca)
accuracy_pca = accuracy_score(y_test, y_pred_pca)

print(f"Training time: {train_time_pca*1000:.2f} ms")
print(f"Test Accuracy: {accuracy_pca*100:.2f}%")
print(f"Number of features: 2")
print(f"Model coefficients shape: {model_pca.coef_.shape}")
print(f"Variance explained: {pca_2.explained_variance_ratio_.sum()*100:.2f}%")

# COMPARISON
print("\n\n" + "=" * 80)
print("📊 COMPARISON RESULTS")
print("=" * 80)

comparison_df = pd.DataFrame({
    'Metric': ['Training Time (ms)', 'Test Accuracy (%)', 'Features Used', 'Variance Captured (%)'],
    'All Features': [f"{train_time_full*1000:.2f}", f"{accuracy_full*100:.2f}", "4", "100"],
    'Reduced (PCA 2)': [f"{train_time_pca*1000:.2f}", f"{accuracy_pca*100:.2f}", "2", f"{pca_2.explained_variance_ratio_.sum()*100:.2f}"]
})

print("\n" + comparison_df.to_string(index=False))

speedup = train_time_full / train_time_pca if train_time_pca > 0 else 1
accuracy_diff = accuracy_full - accuracy_pca

print(f"\n💡 Key Findings:")
print(f"   • Speed improvement: {speedup:.1f}x faster")
print(f"   • Accuracy change: {accuracy_diff*100:+.2f}%")
print(f"   • Features reduced: 50% (from 4 to 2)")
print(f"   • Variance preserved: {pca_2.explained_variance_ratio_.sum()*100:.1f}%")

if speedup > 1 and accuracy_diff >= -0.05:
    print(f"\n✅ RECOMMENDATION: Use reduced features!")
    print(f"   Faster training with minimal accuracy loss")
else:
    print(f"\n⚠️ TRADE-OFF: Speed vs Accuracy")

# Visualization
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# 1. Accuracy comparison
methods = ['All Features\n(4D)', 'Reduced PCA\n(2D)']
accuracies = [accuracy_full*100, accuracy_pca*100]
colors_acc = ['steelblue', 'green']

axes[0, 0].bar(methods, accuracies, color=colors_acc, alpha=0.7, edgecolor='black')
axes[0, 0].set_ylabel('Accuracy (%)')
axes[0, 0].set_title('Model Accuracy Comparison')
axes[0, 0].set_ylim([0, 105])
axes[0, 0].grid(axis='y', alpha=0.3)

for i, v in enumerate(accuracies):
    axes[0, 0].text(i, v+1, f'{v:.1f}%', ha='center', fontweight='bold')

# 2. Training time comparison
times = [train_time_full*1000, train_time_pca*1000]
axes[0, 1].bar(methods, times, color=colors_acc, alpha=0.7, edgecolor='black')
axes[0, 1].set_ylabel('Training Time (ms)')
axes[0, 1].set_title('Training Speed Comparison')
axes[0, 1].grid(axis='y', alpha=0.3)

for i, v in enumerate(times):
    axes[0, 1].text(i, v+0.01, f'{v:.2f}ms', ha='center', fontweight='bold')

# 3. 2D visualization of PCA data
scatter = axes[1, 0].scatter(X_test_pca[:, 0], X_test_pca[:, 1], c=y_test, 
                            cmap='viridis', s=100, alpha=0.6, edgecolors='black')
axes[1, 0].set_xlabel(f'PC1 ({pca_2.explained_variance_ratio_[0]*100:.1f}%)')
axes[1, 0].set_ylabel(f'PC2 ({pca_2.explained_variance_ratio_[1]*100:.1f}%)')
axes[1, 0].set_title('Iris Classes in 2D (After PCA)')
plt.colorbar(scatter, ax=axes[1, 0], label='Iris Species')

# 4. Variance explained
pca_all = PCA()
pca_all.fit(X_train_scaled)
cumsum_var = np.cumsum(pca_all.explained_variance_ratio_)

axes[1, 1].plot(range(1, len(cumsum_var)+1), cumsum_var*100, 'o-', 
               color='darkred', linewidth=2, markersize=8)
axes[1, 1].axhline(y=95, color='green', linestyle='--', label='95% threshold')
axes[1, 1].axvline(x=2, color='orange', linestyle='--', label='Using 2 components')
axes[1, 1].set_xlabel('Number of Components')
axes[1, 1].set_ylabel('Cumulative Variance Explained (%)')
axes[1, 1].set_title('Variance Explained by PCA')
axes[1, 1].legend()
axes[1, 1].grid(alpha=0.3)
axes[1, 1].set_xlim([1, 4])

plt.tight_layout()
plt.show()

print("\n✅ Visualization shows:")
print("   • Similar accuracy with fewer features")
print("   • 2 components capture most variance")
print("   • Clear separation between classes in 2D")
````

## Summary: Dimensionality Reduction Cheat Sheet

### Quick Reference

| Situation | Method | Code | When |
|-----------|--------|------|------|
| **Feature Selection** | SelectKBest | `SelectKBest(f_classif, k=10)` | Supervised |
| **Unsupervised** | PCA | `PCA(n_components=10)` | No labels |
| **Correlation** | Drop High Corr | `corr() > 0.9` | Quick screening |
| **Feature Importance** | Random Forest | `RandomForestClassifier()` | Non-linear |
| **Visualization** | PCA 2D/3D | `PCA(n_components=2)` | Need to visualize |

---

### Decision Flow

```
START: Do you have many features (>50)?
│
├─ NO → Skip dimensionality reduction ✓
│
└─ YES → Step 1: Check correlations
        │
        ├─ High correlation (>0.9)? → Remove one feature
        │
        └─ No high correlation? → Step 2: Calculate variance
                                 │
                                 ├─ Low variance? → Remove feature
                                 │
                                 └─ All have variance? → Step 3: Select Method
                                                        │
                                                        ├─ Supervised task? → Random Forest
                                                        │
                                                        └─ Unsupervised? → PCA
                                                                         │
                                                                         ├─ Need 2D plot? → PCA(2)
                                                                         └─ Otherwise? → PCA(95% var)
                                                                                       │
                                                                                       → DONE ✓
```

---

### Common Mistakes to Avoid

❌ **Mistake 1**: Applying PCA before standardization
```python
# WRONG
pca = PCA()
pca.fit(df)  # Features at different scales!

# CORRECT
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(df)
pca = PCA()
pca.fit(X_scaled)
```

❌ **Mistake 2**: Not validating performance after reduction
```python
# WRONG
# Reduce features → Train model → Deploy

# CORRECT
# Reduce features → Train model → Compare accuracy → If good, deploy
```

❌ **Mistake 3**: Removing too many features
```python
# WRONG
pca = PCA(n_components=1)  # Only 1 component!

# CORRECT
pca = PCA(n_components=0.95)  # Keep 95% variance
```

❌ **Mistake 4**: Data leakage - fitting scaler on all data
```python
# WRONG
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)  # Fits on train+test!
train, test = split(X_scaled)

# CORRECT
X_train, X_test = split(X)
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)  # Fit on train only
X_test_scaled = scaler.transform(X_test)  # Transform test
```

---

### Key Takeaways

1. **Curse of Dimensionality**: More features ≠ Better models
2. **Start Simple**: Try correlation removal first
3. **Validate**: Always compare before and after
4. **Preserve Information**: Aim for 90-95% variance retained
5. **Balance Trade-offs**: Speed vs Accuracy vs Interpretability
6. **Document**: Record which features you selected and why
