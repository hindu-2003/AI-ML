# LCEL (LangChain Expression Language) 

The pipe operator | connects components:


#### PromptTemplate = plain text prompt

#### ChatPromptTemplate = role-based conversation prompt (modern standard)

#### PromptTemplate generates a single text prompt, while ChatPromptTemplate generates role-based messages for chat-oriented LLMs.


### ChatPromptTemplate

✅ Used for Chat-based models

      (e.g., GPT-4, LLaMA, Mistral, Ollama chat models)

🧠 Idea: Creates structured chat messages with roles.


🔑 Key points

    Produces list of messages
    
    Supports roles: system, human, ai
    
    Better instruction control
    
    Ideal for RAG, agents, multi-turn chat

| Situation              | Recommended        |
| ---------------------- | ------------------ |
| Simple text generation | PromptTemplate     |
| Chatbot                | ChatPromptTemplate |
| RAG pipeline           | ChatPromptTemplate |
| Agents / tools         | ChatPromptTemplate |
| Local LLM (Ollama)     | ChatPromptTemplate |



#### ChatPromptTemplate.from_messages()

        is used in LangChain to create a chat-style prompt made up of multiple messages with roles (system, human, ai).

        It is mainly used with chat-based LLMs (GPT, LLaMA, Mistral, Ollama).


Where messages is a list of:

                Tuples → ("role", "content")

                Or Message objects (SystemMessage, HumanMessage, etc.)

| Role     | Purpose                            |
| -------- | ---------------------------------- |
| `system` | Behavior / rules for the model     |
| `human`  | User input                         |
| `ai`     | Example model responses (few-shot) |


```python

from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# Step 1: Create the prompt

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant"),
    ("human", "{question}")
])

# Step 2: Create the LLM

llm = ChatOllama(model="llama3.2")

# Step 3: Create the output parser

parser = StrOutputParser()

# Step 4: Chain them together with | (pipe)

chain = prompt | llm | parser

# Step 5: Invoke the chain

result = chain.invoke({"question": "What is Python?"})
print(result)
```

### diagram


Input → Prompt → LLM → Parser → Output
{"question": "What is Python?"}
          ↓
    "System: You are helpful\nHuman: What is Python?"
          ↓
    [LLM processes]
          ↓
    Message(content="Python is a programming language...")
          ↓
    "Python is a programming language..."


# Simpl Template Chain

```python
from langchain_ollama import OllamaLLM
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser

# Create template
template = "You are a {role}. Answer: {question}"
prompt = PromptTemplate(
    template=template,
    input_variables=["role", "question"]
)

# Create LLM
llm = OllamaLLM(model="llama3.2")

# Create chain
chain = prompt | llm | StrOutputParser()

# Use it
result = chain.invoke({
    "role": "math teacher",
    "question": "What is algebra?"
})
print(result)
```

# Chain (Without Parser)

```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_template("Explain {topic} simply")
llm = ChatOllama(model="llama3.2")

# Chain without parser
chain = prompt | llm

# Invoke
response = chain.invoke({"topic": "quantum physics"})

print(response.content)  # Need .content because no parser
```

# Translation Chain

```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# Translation chain
translation_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a translator. Translate to {language}."),
    ("human", "{text}")
])

translator_chain = (
    translation_prompt 
    | ChatOllama(model="mistral") 
    | StrOutputParser()
)

result = translator_chain.invoke({
    "language": "Hindi",
    "text": "Hello, how are you?"
})
print(result)  
```

# Code Helper Chain

```python
code_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are an expert {language} programmer."),
    ("human", "Write code to: {task}")
])

code_chain = (
    code_prompt 
    | ChatOllama(model="llama3.2", temperature=0.0) 
    | StrOutputParser()
)

code = code_chain.invoke({
    "language": "Python",
    "task": "read a CSV file using pandas"
})
print(code)
```

# Multi-Step Chain

```python
# Step 1: Summarize
summary_prompt = ChatPromptTemplate.from_template(
    "Summarize this in one sentence: {text}"
)

# Step 2: Analyze sentiment
sentiment_prompt = ChatPromptTemplate.from_template(
    "What is the sentiment (positive/negative/neutral) of: {summary}"
)

llm = ChatOllama(model="llama3.2")

# Chain them together
full_chain = (
    {"text": lambda x: x["text"]} 
    | summary_prompt 
    | llm 
    | StrOutputParser()
    | (lambda summary: {"summary": summary})
    | sentiment_prompt
    | llm
    | StrOutputParser()
)

result = full_chain.invoke({
    "text": "I absolutely loved this movie! It was incredible and amazing!"
})
print(result)  # "positive"
```
