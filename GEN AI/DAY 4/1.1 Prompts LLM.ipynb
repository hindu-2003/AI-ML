```python

from langchain_ollama import ChatOllama, OllamaLLM

from langchain_core.prompts import PromptTemplate, ChatPromptTemplate

from langchain_core.output_parsers import StrOutputParser

from langchain_core.messages import SystemMessage, HumanMessage
```

```python

llm = OllamaLLM(model="llama3.2", temperature=0.7)

response = llm.invoke("What is Python?")

print(f"\nBasic Response:\n{response}\n")
```

# Prompt Templates


PromptTemplate(...)

    * This is a class constructor
    
    * It creates a prompt object

input_variables=["question"]

    *  This is a list of variable names

    * The variable name must match what’s inside {} in the template.

template=template

    * This is your actual prompt string
    

```python
# Create a prompt template

template = """You are a helpful assistant.

Answer the following question:

Question: {question}
"""
```

```python

prompt = PromptTemplate(
    input_variables=["question"],
    template=template
)
```

```python
# Format the prompt

formatted_prompt = prompt.format(question="What is machine learning?")

print(f"\nFormatted Prompt:\n{formatted_prompt}")
```

```python
# Use with LLM

response = llm.invoke(formatted_prompt)

print(f"Response:\n{response}\n")
```

# Chain - Prompt + LLM + Parser

```python
# Create a simple chain using LCEL (LangChain Expression Language)

chain = prompt | llm | StrOutputParser()

# Invoke the chain
result = chain.invoke({"question": "What is deep learning?"})

print(f"\nChain Result:\n{result}\n")
```


# ChatPromptTemplate


```python
# Initialize Chat Model

chat_model = ChatOllama(model="llama3.2", temperature=0.7)
```

```python
# Create chat prompt template

chat_prompt = ChatPromptTemplate.from_messages([
    
    ("system", "You are an expert {role}. Always provide detailed and accurate information."),
    
    ("human", "{user_input}")
    
])
```

```python
# Format and invoke

messages = chat_prompt.format_messages(
    role="Python developer",
    user_input="How do I handle exceptions in Python?"
)
```

```python

response = chat_model.invoke(messages)

print(f"\nChat Response:\n{response.content}\n")
```

# Multi-Variable Templates

```python
# Template with multiple variables

multi_template = """You are a {profession} with {years} years of experience.

Task: {task}

Context: {context}

Provide a professional response:"""

multi_prompt = PromptTemplate(
    input_variables=["profession", "years", "task", "context"],
    template=multi_template
)

# Create chain
multi_chain = multi_prompt | llm | StrOutputParser()

# Invoke with multiple inputs
result = multi_chain.invoke({
    "profession": "data scientist",
    "years": "10",
    "task": "Explain gradient descent",
    "context": "Teaching to beginners"
})

print(f"\nMulti-variable Response:\n{result}\n")
```

# Few-Shot Prompting

```python

few_shot_template = """You are a sentiment analyzer. Classify the sentiment as Positive, Negative, or Neutral.

Examples:
Text: "I love this product!"
Sentiment: Positive

Text: "This is the worst experience ever."
Sentiment: Negative

Text: "The package arrived on time."
Sentiment: Neutral

Now classify:
Text: "{text}"
Sentiment:"""

few_shot_prompt = PromptTemplate(
    input_variables=["text"],
    template=few_shot_template
)

chain = few_shot_prompt | llm | StrOutputParser()

# Test few-shot learning
test_texts = [
    "This movie was amazing!",
    "I'm disappointed with the service.",
    "The product works as described."
]

for text in test_texts:
    sentiment = chain.invoke({"text": text})
    print(f"Text: {text}")
    print(f"Sentiment: {sentiment.strip()}\n")

```

# Conversational Chain

```python

conversation_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a friendly AI assistant. Keep responses concise."),
    ("human", "{input}")
])

conversation_chain = conversation_prompt | chat_model | StrOutputParser()

# Simulate conversation
conversation = [
    "Hi, what's your name?",
    "Can you help me learn Python?",
    "What should I learn first?"
]

for message in conversation:
    print(f"User: {message}")
    response = conversation_chain.invoke({"input": message})
    print(f"AI: {response}\n")
```

# Role-Based Prompting

```python

role_prompt = ChatPromptTemplate.from_messages([
    ("system", """You are acting as a {role}. 
    
Your characteristics:
- {trait1}
- {trait2}
- {trait3}

Respond in character to the following:"""),
    ("human", "{query}")
])

role_chain = role_prompt | chat_model | StrOutputParser()

# Example: Tech support role
response = role_chain.invoke({
    "role": "senior technical support engineer",
    "trait1": "Patient and understanding",
    "trait2": "Explains technical concepts simply",
    "trait3": "Always provides step-by-step solutions",
    "query": "My computer won't start. What should I do?"
})

print(f"Response:\n{response}\n")
```

# Output Formatting

```python

structured_template = """Analyze the following text and provide a structured response.

Text: {text}

Provide your analysis in this exact format:
Topic: [main topic]
Sentiment: [positive/negative/neutral]
Key Points: [list 3 key points]
Summary: [one sentence summary]"""

structured_prompt = PromptTemplate(
    input_variables=["text"],
    template=structured_template
)

structured_chain = structured_prompt | llm | StrOutputParser()

analysis_text = "LangChain is an amazing framework for building LLM applications. It makes it easy to chain together prompts, models, and parsers. Developers love its flexibility and extensive documentation."

result = structured_chain.invoke({"text": analysis_text})
print(f"Structured Analysis:\n{result}\n")
```

# What's the difference between OllamaLLM and ChatOllama?

### When would you use temperature=0.0 vs temperature=0.9?


Temperature controls randomness and creativity in the model's responses.

❄️ Temperature = 0.0 (Cold/Deterministic)

When to use:

    ✅ Factual questions
    ✅ Code generation
    ✅ Math problems
    ✅ Data extraction
    ✅ Translation
    ✅ Classification tasks
    ✅ When you need consistent, repeatable answers

Characteristics:

    Always picks the most likely next word
    Same input → Same output (mostly)
    Focused, precise, conservative
    No creativity or variation

🔥 Temperature = 0.9-1.0 (Hot/Creative)

When to use:

    ✅ Creative writing (stories, poems)
    ✅ Brainstorming ideas
    ✅ Marketing copy variations
    ✅ Casual conversation
    ✅ Generating multiple alternatives
    ✅ When you want diverse, unexpected responses


Characteristics:

    Considers less likely word choices
    Same input → Different outputs each time
    More creative, diverse, surprising
    Can be less accurate/factual

🌤️ Temperature = 0.4-0.7 (Balanced) ⭐ Most Common

When to use:

    ✅ General conversation
    ✅ Explanations
    ✅ Product descriptions
    ✅ Email writing
    ✅ Most everyday tasks


Characteristics:

    Good balance of accuracy and variety
    Reliable but not boring
    Default choice for most applications
