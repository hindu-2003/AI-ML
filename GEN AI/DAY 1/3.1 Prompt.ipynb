### Understanding User Prompts vs System Prompts


<b> A prompt is the instruction or input you give to an AI model to get a response. </b>



The quality of the AI’s answer depends heavily on:

    Clarity
    
    Specificity
    
    Context
    
    Constraints



What is a <b>User Prompt</b>?

    A user prompt is the direct instruction or question you send to the AI model. 

    It's what the end user asks for.

What is a <b>System Prompt</b>?

    A system prompt (or system instruction) is a hidden instruction that sets the behavior, tone, and role of the AI.

    It shapes how the AI responds to user prompts.

A <b>User Instruction</b> is:

A<b> System Instruction </b>defines:

    The behavior
    
    The role
    
    The tone
    
    The rules
    
    The limitations

It is set by the developer and not visible to the end user.

It controls how the model behaves.

```python

model = genai.GenerativeModel(
    model_name="gemini-2.5-flash",
    system_instruction="""
    You are a friendly AI teacher.
    Explain concepts in very simple language.
    Use short sentences.
    Give one real-life example.
    Keep the answer under 120 words.
    """
)

response = model.generate_content("Explain AI.")
print(response.text)
```

### Instructional Prompt

Tell the model to do something clearly.

An instructional prompt is when you tell the model exactly what task to do using a direct command or request.

    No examples (unlike few-shot).
    
    Just clear instructions.
    

```python

model.generate_content("Summarize the following paragraph in one sentence:\n\n[The Madhya Pradesh Pollution Control Board has confirmed that officials had incinerated 337 tonnes of toxic wastemoved to a private waste treatment facility in Pithampur from the defunct Union Carbide facility in Bhopal. The event closes a single, but important, chapter in the sordid history of the 1984 Bhopal disaster, the state response to which pushed the city and its people to the edge. The successful incineration is illustrative in that it demanded interventions from the Madhya Pradesh High Court, the Supreme Court of India, and the Union Environment Ministry, among other stakeholders, over more than a decade. In the end, the State government was able to arrange for the waste to be safely disposed of — including an outreach exercise to assuage public anxiety over the emissions — within six months at a cost of ₹126 crore. Evidently the means have always existed; the political initiative to effect them has been wanting. This is an important detail: waste, once it has entered the environment, has a tendency to be converted to different forms; it seldom goes away. The toxic waste incinerated thus far has yielded more than 800 tonnes of ash and residue that officials will have to landfill in a scientific manner. Like solid waste landfills around the country, this new facility will require regular upkeep, monitoring, and funds of its own. The site of the Union Carbide plant also retains several more tonnes of contaminated soil and other hazardous artefacts, plus contaminated subsurface resources in the area.]")
```

### Few-shot Prompting

Provide examples to guide the model.
    
    The model is given a few examples of input-output pairs in the prompt.
    
    This helps it learn the task style and pattern before answering.


### One-Shot Learning (Prompting)

A special case of few-shot prompting where you provide exactly one example of the task.

It helps the model understand the expected format/output.


```python

prompt = """Convert the sentence to past tense:

Example: She goes to school. → She went to school.

Now: He eats an apple.
"""
model.generate_content(prompt)
```

### Zero-shot Prompting

Just ask directly without examples.

    The model is asked to perform a task without seeing any examples in the prompt.

    It relies only on its pre-trained knowledge.

    You just give the instruction, and the model figures it out.
    

```python

model.generate_content("Translate 'How are you?' into Hindi.")
```

| Method        | Examples Given in Prompt | When Used                                       |
| ------------- | ------------------------ | ----------------------------------------------- |
| **Zero-Shot** | 0                        | When model is strong enough / task is simple    |
| **One-Shot**  | 1                        | When format/pattern matters, but task is simple |
| **Few-Shot**  | 2–10                     | When task is complex and model needs guidance   |



Chain-of-Thought (CoT), where the model is encouraged to reason step-by-step before giving the final answer.

Why CoT is Useful

It helps the model:

    Break complex problems into smaller steps
    
    Improve logical reasoning
    
    Increase accuracy in tasks like math, coding, and reasoning.
    

```python
import google.generativeai as genai

prompt = """
You are a careful reasoning assistant.

Think through the solution step by step internally.
Provide:
1) A short explanation (max 3 lines)
2) Final answer clearly labeled

Question:
A laptop costs ₹50,000.
A 10% discount is applied.
Then 18% GST is added.
 Give me step by step reasoning?
What is the final price?
"""

response = model.generate_content(prompt)

print(response.text)
```
