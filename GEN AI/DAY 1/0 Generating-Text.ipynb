### Create a chat session to preserve history

```python

chat = model.start_chat(history=[])
```

start_chat() creates a conversation object (chat).

    The optional history=[] parameter means you’re starting fresh with no prior messages.

    Behind the scenes, Gemini will now keep track of all future exchanges in this session until you reset it

#### Function to interact with Gemini chat

```python

def chat_with_gemini(user_input):
    
    response = chat.send_message(user_input)
    
    print("User:", user_input)
    print("Gemini:", response.text)
    
    return response.text
```

chat.send_message(user_input)

    Adds your message to the conversation history.
    
    Sends it to Gemini.
    
    Returns a response object.

response.text is the model’s answer as plain text.

The function also prints both sides of the conversation (nice for debugging / CLI).

```python
# First message

chat_with_gemini("Hello! What can you do?")
```

```python
# Second message (chat history is remembered)

chat_with_gemini("Tell me more about your capabilities in education.")
```

### Streaming response for single prompt

| Setting        | Behavior                            | Example                                      |
| -------------- | ----------------------------------- | -------------------------------------------- |
| `stream=False` | Wait for full output                | Best for backend scripts or batch processing |
| `stream=True`  | Receive words as they are generated | Best for chat apps or interactive UIs        |


stream=False → get complete model response after generation.

stream=True → get partial responses in real time (token streaming).

```python
def stream_gemini_response(prompt):
    
    response_stream = model.generate_content(prompt, stream=True)
    print("Gemini (streaming):", end=" ", flush=True)
    
    for chunk in response_stream:
        print(chunk.text, end="", flush=True)

stream_gemini_response("Explain how Large Language Models understand text.")
```

stream=True yields chunks as they arrive.

chunk.text is the partial text for that chunk.

### Streaming with chat (multi-turn)

```python
chat = model.start_chat(history=[])

def stream_chat_response(user_input):
    print("User:", user_input)
    print("Gemini (streaming):", end=" ", flush=True)
    response_stream = chat.send_message(user_input, stream=True)
    
    for chunk in response_stream:
        print(chunk.text, end="", flush=True)
    print()

stream_chat_response("Hi! What can you do?")
stream_chat_response("Give me a summary of generative AI.")
```
