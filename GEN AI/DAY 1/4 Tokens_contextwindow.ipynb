What are Tokens?

A token is a small unit of text (a word or part of a word). 

Every prompt and response uses tokens, which affects:

    Cost: You pay per token used
    
    Speed: Fewer tokens = faster response
    
    Quality: Sometimes shorter responses are better, sometimes you need more detail


🔑 Why Tokens Matter: The Economics

Tokens are the CURRENCY of AI.

1. You pay per token (input + output)

    Every word you send costs money
    Every word the AI generates costs money

2. Different models have different prices

3. One word ≠ One token

    "Explain" = 1 token

    "database" = 1 token... but uncommon words split into 2-3 tokens

    Punctuation, numbers, special characters all count

5. Context windows are limited

    Every prompt you send uses up your available "memory"

    Long prompts = less space for conversation

    Less space = worse context awareness = worse answers


<b>o200k_base</b> is a tokenizer encoding vocabulary used by OpenAI’s newer models (like GPT-4o family).

What is o200k_base?

o200k_base is a tokenization scheme with about 200,000 tokens in its vocabulary.

It is designed to efficiently encode multiple languages, code, and symbols.

    o → OpenAI
    
    200k → ~200,000 token vocabulary
    
    base → base encoding scheme used for tokenization
    

```python
import tiktoken

encoding = tiktoken.get_encoding("o200k_base")

tokens = encoding.encode("Hi my name is Ganapathi Raju and I love to teach")

print(tokens)
print("Token count:", len(tokens))
```

```python
for token_id in tokens:
    token_text = encoding.decode([token_id])
    print(f"{token_id} = {token_text}")
```

### Context Window

The <b>context window</b> is the maximum amount of text (measured in tokens, not words) that a language model can read and remember at one time.

It includes:
    System prompt
    User prompt
    Chat history
    Retrieved documents (in RAG)
    Model response

All of this together must fit inside the model’s context window size.


input_token_limit → Context window size

output_token_limit → Maximum tokens the model can generate


For Gemini 2.5 Flash:

    Input ≈ 1,048,576 tokens
    
    Output ≈ 65,536 tokens
