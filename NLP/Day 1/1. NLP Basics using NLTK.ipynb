### POS Tagging

The process of classifying words into their parts of speech and labeling them accordingly is known as part-of-speech tagging, <b>POS-tagging</b> , or simply tagging. 

Parts of speech are also known as word classes or lexical categories. 

The collection of tags used for a particular task is known as a tagset

### WSD

WSD (Word Sense Disambiguation) in NLTK is about figuring out the correct meaning (sense) of a word based on its context.

For example, the word “bank” can mean:

A financial institution 🏦

The side of a river 🌊

WSD helps the system choose the right meaning depending on the sentence.

```python
from nltk.wsd import lesk

from nltk.tokenize import word_tokenize

sentence = "I went to the bank to deposit money"
words = word_tokenize(sentence)

sense = lesk(words, "bank")

print(sense)
print(sense.definition())
```

```python
sentence = "The fisherman sat on the bank of the river"
words = word_tokenize(sentence)

sense = lesk(words, "bank")
print(sense.definition())
```

### NER

NER (Named Entity Recognition) is an NLP task that identifies and classifies real-world entities in text into predefined categories like:

Person, Organization, Location, Date, Time, Money, Medical terms, etc.

Dr. Raju works at AIIMS Delhi since 2022.

Dr. Raju → PERSON

AIIMS Delhi → ORGANIZATION

2022 → DATE

Delhi → LOCATION

```python
import nltk
nltk.download('maxent_ne_chunker')
nltk.download('words')
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')
```

```python
text = "Elon Musk is the CEO of Tesla in California"

tokens = nltk.word_tokenize(text)
pos_tags = nltk.pos_tag(tokens)

entities = nltk.ne_chunk(pos_tags)

print(entities)
```
