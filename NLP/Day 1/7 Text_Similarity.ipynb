```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

def preprocess_text(text):
   
    stop_words = set(stopwords.words('english'))
    stemmer = PorterStemmer()
    tokens = nltk.word_tokenize(text.lower())
    filtered_tokens = [stemmer.stem(token) for token in tokens if token not in stop_words]
    return ' '.join(filtered_tokens)

def calculate_plagiarism(doc1, doc2, threshold=0.8):
   
    preprocessed_doc1 = preprocess_text(doc1)
    preprocessed_doc2 = preprocess_text(doc2)

    vectorizer = TfidfVectorizer()
    tfidf_matrix = vectorizer.fit_transform([preprocessed_doc1, preprocessed_doc2])
    cosine_sim = cosine_similarity(tfidf_matrix)[0][1]

    if cosine_sim >= threshold:
        return True
    else:
        return False

doc1 = "This is the first document. It contains some unique content."
doc2 = "This is the first document. It contains some unique content"
doc3 = "This is a completely different document."

print(f"Doc1 and Doc2: {calculate_plagiarism(doc1, doc2)}")  # Likely to detect plagiarism
print(f"Doc1 and Doc3: {calculate_plagiarism(doc1, doc3)}")  # Unlikely to detect plagiarism
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity, euclidean_distances, manhattan_distances
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

def preprocess_text(text):
    """
    Preprocesses the text by removing stop words and stemming.

    Args:
        text: The input text.

    Returns:
        The preprocessed text.
    """
    stop_words = set(stopwords.words('english'))
    stemmer = PorterStemmer()
    tokens = nltk.word_tokenize(text.lower())
    filtered_tokens = [stemmer.stem(token) for token in tokens if token not in stop_words]
    return ' '.join(filtered_tokens)

def calculate_similarity(doc1, doc2, metric='cosine', threshold=0.8):
    """
    Calculates the similarity between two documents using different metrics.

    Args:
        doc1: The first document.
        doc2: The second document.
        metric: The similarity metric to use ('cosine', 'euclidean', 'manhattan', 'jaccard').
        threshold: The similarity threshold.

    Returns:
        True if similarity exceeds the threshold, False otherwise.
    """
    preprocessed_doc1 = preprocess_text(doc1)
    preprocessed_doc2 = preprocess_text(doc2)

    vectorizer = TfidfVectorizer()
    tfidf_matrix = vectorizer.fit_transform([preprocessed_doc1, preprocessed_doc2])

    if metric == 'cosine':
        sim_score = cosine_similarity(tfidf_matrix)[0][1]
    elif metric == 'euclidean':
        sim_score = 1 - euclidean_distances(tfidf_matrix)[0][1]  # Invert Euclidean distance for similarity
    elif metric == 'manhattan':
        sim_score = 1 - manhattan_distances(tfidf_matrix)[0][1]  # Invert Manhattan distance for similarity
    else:
        raise ValueError("Invalid similarity metric. Choose from 'cosine', 'euclidean', 'manhattan', 'jaccard'.")

    return sim_score >= threshold

# Example usage
doc1 = "This is the first document. It contains some unique content."
doc2 = "This is the first document. It contains some unique content."
doc3 = "This is a completely different document."

metrics = ['cosine', 'euclidean', 'manhattan']
for metric in metrics:
    print(f"Doc1 and Doc2 ({metric}): {calculate_similarity(doc1, doc2, metric)}")
    print(f"Doc1 and Doc3 ({metric}): {calculate_similarity(doc1, doc3, metric)}")
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.tokenize import word_tokenize

def preprocess_text(text):
   
    stop_words = set(stopwords.words('english'))
    stemmer = PorterStemmer()
    tokens = word_tokenize(text.lower())
    filtered_tokens = [stemmer.stem(word) for word in tokens if word not in stop_words]
    return " ".join(filtered_tokens)

def search_documents(query, documents):
   
    preprocessed_query = preprocess_text(query)
    results = []
    for i, doc in enumerate(documents):
        if preprocessed_query in preprocess_text(doc):
            results.append(i)
    return results

# Example usage
documents = [
    "This is the first document.",
    "This is the second document.",
    "The first document contains some important information.",
    "This document is completely different."
]

query = "first document"

result_indices = search_documents(query, documents)

print("Matching documents:")
for i in result_indices:
    print(f"Document {i+1}: {documents[i]}")
```

```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize

def preprocess_text(text):
 
    stop_words = set(stopwords.words('english'))
    lemmatizer = WordNetLemmatizer()
    tokens = word_tokenize(text.lower())
    filtered_tokens = [lemmatizer.lemmatize(word) for word in tokens if word not in stop_words]
    return " ".join(filtered_tokens)

def search_documents(query, documents):
   
    preprocessed_query = preprocess_text(query)
    query_tokens = set(preprocessed_query.split())  # Convert to set for faster matching

    results = []
    for i, doc in enumerate(documents):
        preprocessed_doc = preprocess_text(doc)
        doc_tokens = set(preprocessed_doc.split())

        # Check if all query tokens are present in the document
        if query_tokens.issubset(doc_tokens):
            results.append(i)

    return results

# Example usage
documents = [
    "This is the first document.",
    "This is the second document.",
    "The first document contains some important information.",
    "This document is completely different."
]

query = "first document"

result_indices = search_documents(query, documents)

print("Matching documents:")
for i in result_indices:
    print(f"Document {i+1}: {documents[i]}")
```
