
<b>Natural Language Processing (NLP)</b> 

    is a field of Artificial Intelligence (AI) that enables computers to understand, interpret, and generate human language.

It combines linguistics, 
            
            computer science, and 
            
            machine learning 
            
            to help machines interact with humans using natural language (English, Hindi) instead of programming code.

<b>Why NLP is Important</b>

NLP is behind many applications we use daily:

    Voice Assistants: Siri, Alexa, Google Assistant
    
    Chatbots: Customer support chat systems
    
    Machine Translation: Google Translate
    
    Sentiment Analysis: Social media monitoring
    
    Search Engines: Understanding your search intent

<b>Main Types of NLP Tasks</b>

NLP can be divided into two main types (based on purpose) and then further broken down into core tasks.

<b>1. Natural Language Understanding (NLU) </b>

         Helps machines comprehend human language.

    Key NLU tasks:

        Tokenization: Splitting text into words or sentences.
            Example: "I love NLP" → ["I", "love", "NLP"]
        
        Part-of-Speech Tagging (POS): Identifying word roles (noun, verb, etc.)
        
        Named Entity Recognition (NER): Detecting names, locations, dates, etc.
            Example: "Barack Obama was born in Hawaii" → [Person: Barack Obama, Location: Hawaii]
        
        Sentiment Analysis: Understanding emotions or opinions.
        
        Semantic Analysis: Understanding the meaning behind words.

<b>2. Natural Language Generation (NLG) </b>

    Helps machines produce human-like language.

    Key NLG tasks:

        Text Summarization: Shortening long documents while keeping meaning.
        
        Machine Translation: Translating text between languages.
        
        Question Answering: Providing direct answers from text.
        
        Chatbot Conversations: Generating human-like replies.
        
        Report Generation: Automatically creating summaries or insights from data.

<b> Common NLP Applications </b>

    Text Classification: Spam detection, topic labeling
    
    Information Extraction: Pulling facts from text
    
    Speech Recognition: Converting speech → text
    
    Conversational AI: Intelligent assistants
    
    Document Summarization: Auto-summarizing reports
    
    Text-to-Speech (TTS): Reading out text naturally

<b>Levels of NLP</b>

    NLP operates at several linguistic levels, each handling a different layer of meaning and structure.

| **Level**                                  | **Description**                                                                                   | **Example**                                                                      |
| ------------------------------------------ | ------------------------------------------------------------------------------------------------- | -------------------------------------------------------------------------------- |
| 1️** Phonological / Morphological Level** | Deals with sounds (phonology) and word formation (morphology).                                    | “Unhappiness” → *un* (prefix) + *happy* (root) + *ness* (suffix)                 |
| 2️ **Lexical Level**                      | Deals with **individual words** and their meanings (dictionary-level).                            | “Bank” → can mean *river bank* or *financial institution*                        |
| 3️ **Syntactic Level**                    | Focuses on **grammar and sentence structure** — how words combine to form valid sentences.        | “The cat sat on the mat” ✅ vs “Cat the on sat mat the” ❌                         |
| 4️ **Semantic Level**                     | Deals with **meaning of words and sentences** — what the sentence *means*.                        | “He unlocked the door” → understanding the *action* of opening a locked door.    |
| 5️ **Pragmatic Level**                    | Interprets **contextual or situational meaning** — what the speaker *intends*.                    | “Can you open the window?” → actually a *request*, not a question about ability. |
| 6️ **Discourse Level**                    | Considers the meaning of **a group of sentences** — coherence across a paragraph or conversation. | “John bought a car. He loves it.” → “He” refers to *John*.                       |
| 7️ **World Knowledge Level**              | Uses **real-world facts or commonsense knowledge** to interpret text correctly.                   | “The ice melted because it was hot” → knowing *heat melts ice*.                  |


Text Input
   ↓
1. Phonological / Morphological → Word formation
   ↓
2. Lexical → Word meaning
   ↓
3. Syntactic → Grammar / structure
   ↓
4. Semantic → Sentence meaning
   ↓
5. Pragmatic → Contextual meaning
   ↓
6. Discourse → Linking sentences
   ↓
7. World Knowledge → Real-world understanding


<b> Common Text Preprocessing Techniques </b>

🧹 1️⃣ Lowercasing

Convert all text to lowercase so that words like “Python” and “python” are treated the same.

text = text.lower()

Example:
        👉 "NLP Is Fun" → "nlp is fun"

```python
import string
print(string.punctuation)
```

✂️ 2️⃣ Removing Punctuation, Numbers, and Special Characters

These often don’t add value for text understanding.

import re
text = re.sub(r'[^a-zA-Z\s]', '', text)

Example:
👉 "Hello!!! Are you #1?" → "Hello Are you"

🧾 3️⃣ Tokenization

Splitting text into individual words or sentences.

from nltk.tokenize import word_tokenize
tokens = word_tokenize(text)


Example: 👉 "NLP is fun" → ['NLP', 'is', 'fun']

🧺 4️⃣ Removing Stopwords

Stopwords are common words (like is, the, a, an, of) that don’t add significant meaning.

from nltk.corpus import stopwords
filtered = [w for w in tokens if w not in stopwords.words('english')]


Example: 👉 "This is a good example" → ['good', 'example']

🧩 5️⃣ Stemming

Reduce words to their root form by chopping off endings.

from nltk.stem import PorterStemmer
ps = PorterStemmer()
ps.stem('running') → 'run'


Example:
👉 "playing", "played", "plays" → "play"

⚠️ Stemming is rule-based and may not always produce real words.

🪶 6️⃣ Lemmatization

More advanced than stemming — uses vocabulary and grammar rules to get the base word (lemma).

from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
lemmatizer.lemmatize('better', pos='a') → 'good'


Example:
👉 "studies" → "study", "better" → "good"

🧼 7️⃣ Removing Extra Whitespaces

Clean up unnecessary spaces.

text = ' '.join(text.split())

🔠 8️⃣ Handling Contractions

Expand short forms (like “don’t” → “do not”, “I’m” → “I am”).
Useful for sentiment analysis.

import contractions
text = contractions.fix("I'm learning NLP")

🧰 9️⃣ Handling Numbers

Depending on the context:

Remove numbers: if not relevant

Convert to text: if meaning matters (e.g., “100” → “one hundred”)

🧍‍♂️ 10️⃣ Part-of-Speech (POS) Tagging

Assign grammatical roles to each word — helps in lemmatization and entity recognition.

import nltk
nltk.pos_tag(['NLP', 'is', 'fun'])

🧠 11️⃣ Named Entity Recognition (NER)

Identify names, places, organizations, dates, etc.

Example:
👉 “Elon Musk founded SpaceX in 2002.”
→ [Person: Elon Musk], [Organization: SpaceX], [Date: 2002]

🧮 12️⃣ Text Vectorization 

Convert cleaned text into numerical form for ML models.

| **Method**             | **Description**                                |
| ---------------------- | ---------------------------------------------- |
| **Bag of Words (BoW)** | Counts occurrences of each word                |
| **TF-IDF**             | Weights words based on importance              |
| **Word2Vec / GloVe**   | Creates semantic word embeddings               |
| **BERT embeddings**    | Contextualized, deep learning-based embeddings |
