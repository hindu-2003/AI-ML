```python
corpus="""Data science is the study of data to extract meaningful insights for business. It is a multidisciplinary approach that combines principles and practices from the fields of mathematics, statistics, artificial intelligence, and computer engineering to analyze large amounts of data.
"""
```

```python
print(corpus)
```

```python
##  Tokenization
## Sentence-->paragraphs
from nltk.tokenize import sent_tokenize
```

```python
documents=sent_tokenize(corpus)
```

```python
documents
```

```python
type(documents)
```

```python
for sentence in documents:
    print(sentence)
```

```python
from nltk.tokenize import sent_tokenize
## Tokenization 
## Paragraph-->words
## sentence--->words
from nltk.tokenize import word_tokenize

from nltk.tokenize import wordpunct_tokenize
```

```python
print(word_tokenize(corpus))
```

```python
for sentence in documents:
    print(word_tokenize(sentence))
```

```python
s = '''Good muffins cost $3.88\nin New York.  
       Please buy me ... two of them.\n\nThanks.'''

print(wordpunct_tokenize(s))
```

```python
wordpunct_tokenize(corpus)
```
