```python
from sklearn.decomposition import TruncatedSVD
from sklearn.feature_extraction.text import TfidfVectorizer

documents = [
    "I love machine learning and AI.",
    "Deep learning is a subset of machine learning.",
    "Natural Language Processing is part of AI.",
    "AI and machine learning are the future."
]
```

```python
# Step 1: Convert documents to a term-document matrix
```

```python
vectorizer = TfidfVectorizer(stop_words='english')
X = vectorizer.fit_transform(documents)
X.shape
X
```

```python
# Step 2: Apply LSA (TruncatedSVD)
```

```python

lsa = TruncatedSVD(n_components=3, n_iter=100)
lsa_topic_matrix = lsa.fit_transform(X)
```

```python
lsa_topic_matrix
```

```python
lsa.components_
```

```python
# Step 3: Display Topics (top terms per topic)
```

```python

terms = vectorizer.get_feature_names_out()

for index, topic in enumerate(lsa.components_):
    print(f"Topic {index + 1}:")
    top_terms_idx = topic.argsort()[-5:][::-1]
    for term_idx in top_terms_idx:
        print(f"{terms[term_idx]}")
    print()
```
