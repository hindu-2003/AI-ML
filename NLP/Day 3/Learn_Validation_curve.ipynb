# Learning Curve

A learning curve is a graph that shows how a machine learning model’s performance changes as it learns from more training data.

It helps you understand whether your model is learning effectively or not.

The plot usually has:

        X-axis: Number of training samples (how much data the model has seen).
        
        Y-axis: Model accuracy (or error) on both training and validation data.


Training curve – how well the model performs on the data it was trained on.

Validation curve – how well it performs on unseen data (to test generalization).

| Pattern                                                           | What it means                                         |
| ----------------------------------------------------------------- | ----------------------------------------------------- |
| ✅ **Both training & validation accuracy high and close together** | Good fit – model generalizes well                     |
| ⚠️ **Training high, validation much lower**                       | Overfitting – model memorizes training data           |
| ⚠️ **Both low**                                                   | Underfitting – model too simple or not trained enough |
| 📈 **Validation improves with more data**                         | Adding more data can help the model                   |


Why it’s useful:

    Tells you if your model needs more data.
    
    Helps you detect overfitting or underfitting.
    
    Guides you to tune model complexity (e.g., simpler or more powerful model).

### A learning curve is a diagnostic tool that shows how well your model learns and generalizes as it sees more training data.

# Validation curve

A validation curve is a plot that shows how a model’s performance changes as you vary one specific hyperparameter — while keeping the rest fixed.

It helps you find the best value of a model parameter and check if the model is overfitting or underfitting.

The X-axis – values of the hyperparameter (for example, C in Logistic Regression or max_depth in Decision Trees).

The Y-axis – model accuracy (or error).

| Pattern                      | Meaning           | Action                                                                   |
| ---------------------------- | ----------------- | ------------------------------------------------------------------------ |
| **Training ↑, Validation ↓** | Overfitting       | Model too complex — reduce parameter (e.g., smaller `C`, shallower tree) |
| **Both low**                 | Underfitting      | Model too simple — increase parameter (e.g., larger `C`, deeper tree)    |
| **Validation curve peaks**   | Optimal parameter | Choose the parameter value where validation accuracy is highest          |
