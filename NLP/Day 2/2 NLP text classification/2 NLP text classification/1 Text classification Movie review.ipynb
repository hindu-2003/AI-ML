## Text Classification movie review

### Steps in Text Processing

Step 1) Read in a collection of documents - a *corpus*

Step 2) Transform text into numerical vector data using a pipeline

Step 3) Create a classifier

Step 4) Fit/train the classifier

Step 5)Test the classifier on new data

Step 6) Evaluate performance

Dataset : http://www.cs.cornell.edu/people/pabo/movie-review-data/

```python
import numpy as np
import pandas as pd

df = pd.read_csv('moviereviews.tsv', sep='\t')
df
```

```python
df['review'][0]
```

```python
len(df)
```

```python
df.isnull().sum()
```

```python
df.dropna(inplace=True)
len(df)
```

```python
blanks = []  # start with an empty list
        
for i,lb,rv in df.itertuples():  # iterate over the DataFrame
    if type(rv)==str:            # avoid NaN values
        if rv.isspace():         # test 'review' for whitespace
            blanks.append(i)     # add matching index numbers to the list
        
print(len(blanks), 'blanks: ', blanks)
```

```python
df.drop(blanks, inplace=True)

len(df)
```

```python
2000 - 62
```

```python
df['label'].value_counts()
```

```python
from sklearn.model_selection import train_test_split

X = df['review']
y = df['label']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

```python
X
```

```python
y
```

```python
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import LinearSVC

# Naïve Bayes:
text_clf_nb = Pipeline([('tfidf', TfidfVectorizer()),
                     ('clf', MultinomialNB()),
])

# Linear SVC:
text_clf_lsvc = Pipeline([('tfidf', TfidfVectorizer()),
                     ('clf', LinearSVC()),
])
```

```python
text_clf_nb
```

```python
text_clf_nb.fit(X_train, y_train)
```

```python
# Form a prediction set
predictions = text_clf_nb.predict(X_test)
```

```python
# Report the confusion matrix
from sklearn import metrics
print(metrics.confusion_matrix(y_test,predictions))
```

```python
# Print a classification report
print(metrics.classification_report(y_test,predictions))
```

```python
# Print the overall accuracy
print(metrics.accuracy_score(y_test,predictions))
```

```python
text_clf_lsvc
```

```python
text_clf_lsvc.fit(X_train, y_train)
```

```python
# Form a prediction set
predictions = text_clf_lsvc.predict(X_test)
```

```python
# Report the confusion matrix
from sklearn import metrics
print(metrics.confusion_matrix(y_test,predictions))
```

```python
# Print a classification report
print(metrics.classification_report(y_test,predictions))
```

```python
# Print the overall accuracy
print(metrics.accuracy_score(y_test,predictions))
```
