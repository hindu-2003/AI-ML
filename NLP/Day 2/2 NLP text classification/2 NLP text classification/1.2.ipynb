```python
import numpy as np
import pandas as pd
import nltk
import re
```

```python

#nltk.download('stopwords')
#nltk.download('wordnet')

df = pd.read_csv('moviereviews.tsv', sep='\t')
```

```python
# Remove nulls and non-string reviews

df.dropna(inplace=True)
df = df[df['review'].apply(lambda x: isinstance(x, str))]
```

```python
from sklearn.model_selection import train_test_split

X = df['review']
y = df['label']

X_train, X_test, y_train, y_test = train_test_split( X, y, test_size=0.33, random_state=42)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python

best_pipeline = Pipeline([
    ('tfidf', TfidfVectorizer(
        lowercase=True,
        stop_words=None,
        ngram_range=(1,2),     
        min_df=3,
        max_df=0.9,
        sublinear_tf=True
    )),
    ('clf', LinearSVC())
])
```

```python

# Train model
best_pipeline.fit(X_train, y_train)

# Predict
y_pred = best_pipeline.predict(X_test)
```

```python

print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nConfusion Matrix:\n", confusion_matrix(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

⭐ When should you ALWAYS use sublinear TF?

✔ Sentiment analysis
✔ Toxic comment classification
✔ Movie/book review classification
✔ Any emotional text
✔ Any dataset where people repeat words
✔ When using LinearSVC or Logistic Regression


⭐ What sublinear_tf=True does

It applies a logarithmic scaling:

TF = 1 + log(count)

| Word Count | Standard TF | Sublinear TF     |
| ---------- | ----------- | ---------------- |
| 1          | 1           | 1                |
| 5          | 5           | 1 + log(5) ≈ 2.6 |
| 20         | 20          | 1 + log(20) ≈ 4  |


df = df[df['review'].apply(lambda x: isinstance(x, str))]

Meaning: It keeps only those rows where the value in the column review is a string.

Why do this?

Sometimes your dataset contains:
    NaN
    None
    numbers (int/float)
    lists
    other non-string value
Text preprocessing functions (like .lower(), regex, stemming, etc.) require the text to be a string.
So this line filters the DataFrame so only rows with valid text remain.
