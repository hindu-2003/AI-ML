```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Load the dataset

data = pd.read_csv('advertising.csv')
```

```python
# Display the first few rows of the dataset

print(data)
```

```python
# Check for missing values

print(data.isnull().sum())
```

```python

# Visualize the relationship between features and target variable
plt.figure(figsize=(12, 6))

plt.subplot(1, 3, 1)
plt.scatter(data['TV'], data['sales'])
plt.title('TV vs Sales')

plt.subplot(1, 3, 2)
plt.scatter(data['radio'], data['sales'])
plt.title('Radio vs Sales')

plt.subplot(1, 3, 3)
plt.scatter(data['newspaper'], data['sales'])
plt.title('Newspaper vs Sales')

plt.show()
```

<b>Relationship Between Advertising Channels and Sales</b>


```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))

for i, col in enumerate(['TV', 'radio', 'newspaper'], 1):
    plt.subplot(1, 3, i)
    plt.boxplot(data[col])
    plt.title(col)

plt.show()
```

---

```python
# Define features (X) and target variable (y)

X = data[['TV', 'radio', 'newspaper']]

y = data['sales']
```

```python
X
```

```python
y
```

```python

from sklearn.preprocessing import StandardScaler

# Normalize the features
scaler = StandardScaler()
X = scaler.fit_transform(X)
```

```python
X
```

```python
y
```

```python

from sklearn.model_selection import train_test_split
# Split the data into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
X_train
```

```python
X_test
```

```python

from sklearn.linear_model import LinearRegression

model = LinearRegression()
```

```python
# Train the model

model.fit(X_train, y_train)
```

```python
# Make predictions on the test set

y_pred = model.predict(X_test)
```

```python
y_pred
```

```python
from sklearn.metrics import mean_squared_error, r2_score

# Calculate Mean Squared Error (MSE) and R-squared (R2) score
mse = mean_squared_error(y_test, y_pred)

rmse = np.sqrt(mse)

r2 = r2_score(y_test, y_pred)

print(f'Root Mean Squared Error: {rmse}')
print(f'R-squared: {r2}')
```
